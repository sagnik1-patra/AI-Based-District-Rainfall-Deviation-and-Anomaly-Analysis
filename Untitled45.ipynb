```python
import os
import json
import yaml
import pickle
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    cross_val_score,
    StratifiedKFold
)

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)

from tensorflow.keras.callbacks import EarlyStopping


# ======================================================================
# STEP 2: RANDOM SEED
# ======================================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)


# ======================================================================
# STEP 3: PROJECT PATH
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District Rainfall Deviation and Anomaly Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "Rainfalldistrict_2.csv"
)

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# STEP 4: PSO OUTPUT PATHS
# ======================================================================

PSO_H5_PATH = os.path.join(
    PROJECT_DIR,
    "pso_rainfall_model.h5"
)

PSO_PKL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_rainfall_model.pkl"
)

PSO_YAML_PATH = os.path.join(
    PROJECT_DIR,
    "pso_config.yaml"
)

PSO_JSON_PATH = os.path.join(
    PROJECT_DIR,
    "pso_metadata.json"
)


PSO_ACCURACY_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_accuracy_graph.png"
)

PSO_HEATMAP_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_heatmap.png"
)

PSO_COMPARISON_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_comparison_graph.png"
)


PSO_RESULT_CSV = os.path.join(
    PROJECT_DIR,
    "pso_result.csv"
)

PSO_RESULT_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_result_graph.png"
)


PSO_PREDICTION_CSV = os.path.join(
    PROJECT_DIR,
    "pso_prediction.csv"
)

PSO_PREDICTION_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_prediction_graph.png"
)


PSO_FEATURE_SELECTION_CSV = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection.csv"
)

PSO_FEATURE_SELECTION_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_feature_selection_graph.png"
)

PSO_FEATURE_IMPORTANCE_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_feature_importance_graph.png"
)


PSO_ACTUAL_NORMAL_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_actual_vs_normal_rainfall.png"
)

PSO_CATEGORY_GRAPH = os.path.join(
    PROJECT_DIR,
    "pso_category_distribution.png"
)


# ======================================================================
# STEP 5: START
# ======================================================================

print("=" * 80)

print(
    "AI-BASED DISTRICT RAINFALL DEVIATION AND ANOMALY ANALYSIS"
)

print(
    "PARTICLE SWARM OPTIMIZATION (PSO)"
)

print("=" * 80)


print("\nDataset:")

print(CSV_PATH)


# ======================================================================
# STEP 6: CHECK DATASET
# ======================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )


print("\nDataset found successfully.")


# ======================================================================
# STEP 7: LOAD DATA
# ======================================================================

df = pd.read_csv(
    CSV_PATH
)


print("\nDataset loaded successfully.")

print(
    "Dataset shape:",
    df.shape
)


# ======================================================================
# STEP 8: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        "\n",
        " ",
        regex=False
    )
    .str.replace(
        "\r",
        " ",
        regex=False
    )
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


# ======================================================================
# STEP 9: REMOVE EMPTY DATA
# ======================================================================

df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)

df.reset_index(
    drop=True,
    inplace=True
)


print("\nCleaned shape:")

print(df.shape)


print("\nColumns:")

for i, col in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i}. {col}"
    )


# ======================================================================
# STEP 10: DISTRICT COLUMN
# ======================================================================

district_column = None


for col in df.columns:

    if "district" in col.lower():

        district_column = col

        break


if district_column is None:

    district_column = df.columns[0]


print("\nDistrict column:")

print(district_column)


# ======================================================================
# STEP 11: CLEAN DISTRICT NAMES
# ======================================================================

df[district_column] = (
    df[district_column]
    .astype(str)
    .str.strip()
)


invalid_names = [
    "",
    "nan",
    "none",
    "null"
]


df = df[
    ~df[district_column]
    .str.lower()
    .isin(invalid_names)
].copy()


df.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# STEP 12: CONVERT DATA TO NUMERIC
# ======================================================================

for col in df.columns:

    if col == district_column:

        continue


    cleaned = (
        df[col]
        .astype(str)
        .str.replace(
            ",",
            "",
            regex=False
        )
        .str.strip()
    )


    cleaned = cleaned.replace(

        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None",
            "null"
        ],

        np.nan
    )


    numeric_version = pd.to_numeric(

        cleaned,

        errors="coerce"
    )


    if numeric_version.notna().sum() > 0:

        df[col] = numeric_version


# ======================================================================
# STEP 13: NUMERIC COLUMNS
# ======================================================================

numeric_columns = df.select_dtypes(

    include=[np.number]

).columns.tolist()


print("\nNumerical columns:")

for col in numeric_columns:

    print(col)


if len(numeric_columns) < 2:

    raise ValueError(
        "Not enough numerical rainfall columns."
    )


# ======================================================================
# STEP 14: HANDLE MISSING VALUES
# ======================================================================

for col in numeric_columns:

    median_value = df[col].median()


    if pd.isna(
        median_value
    ):

        median_value = 0


    df[col] = df[col].fillna(
        median_value
    )


# ======================================================================
# STEP 15: IDENTIFY ACTUAL / NORMAL
# ======================================================================

actual_columns = []

normal_columns = []


for col in numeric_columns:

    lower = col.lower()


    if "actual" in lower:

        actual_columns.append(
            col
        )


    elif "normal" in lower:

        normal_columns.append(
            col
        )


# ======================================================================
# STEP 16: FALLBACK
# ======================================================================

if (
    len(actual_columns) == 0
    or
    len(normal_columns) == 0
):

    print(
        "\nActual/Normal column names not detected."
    )

    print(
        "Using alternating numerical columns."
    )


    actual_columns = (
        numeric_columns[0::2]
    )


    normal_columns = (
        numeric_columns[1::2]
    )


# ======================================================================
# STEP 17: MATCH ACTUAL/NORMAL PAIRS
# ======================================================================

pair_count = min(

    len(actual_columns),

    len(normal_columns)
)


if pair_count == 0:

    raise ValueError(
        "Could not identify Actual/Normal rainfall columns."
    )


actual_columns = actual_columns[
    :pair_count
]


normal_columns = normal_columns[
    :pair_count
]


print("\nActual columns:")

for col in actual_columns:

    print(col)


print("\nNormal columns:")

for col in normal_columns:

    print(col)


# ======================================================================
# STEP 18: CALCULATE TOTAL ACTUAL
# ======================================================================

df[
    "Calculated_Total_Actual"
] = df[
    actual_columns
].sum(
    axis=1
)


# ======================================================================
# STEP 19: CALCULATE TOTAL NORMAL
# ======================================================================

df[
    "Calculated_Total_Normal"
] = df[
    normal_columns
].sum(
    axis=1
)


# ======================================================================
# STEP 20: RAINFALL DEVIATION
# ======================================================================

df[
    "Rainfall_Deviation_Percentage"
] = np.where(

    df[
        "Calculated_Total_Normal"
    ] != 0,

    (

        (

            df[
                "Calculated_Total_Actual"
            ]

            -

            df[
                "Calculated_Total_Normal"
            ]

        )

        /

        df[
            "Calculated_Total_Normal"
        ]

    )

    * 100,

    0
)


# ======================================================================
# STEP 21: CREATE TARGET
# ======================================================================

def rainfall_category(
    deviation
):

    if deviation < -20:

        return "Deficient"

    elif deviation <= 20:

        return "Normal"

    else:

        return "Excess"


df[
    "Rainfall_Category"
] = df[
    "Rainfall_Deviation_Percentage"
].apply(
    rainfall_category
)


print("\nCategory distribution:")

print(
    df[
        "Rainfall_Category"
    ].value_counts()
)


# ======================================================================
# STEP 22: ORIGINAL FEATURES
# ======================================================================

feature_columns = list(

    dict.fromkeys(

        actual_columns
        +
        normal_columns
    )
)


X_original = df[
    feature_columns
].copy()


y_text = df[
    "Rainfall_Category"
].copy()


print(
    "\nNumber of original features:",
    len(feature_columns)
)


for feature in feature_columns:

    print(
        " -",
        feature
    )


# ======================================================================
# STEP 23: LABEL ENCODER
# ======================================================================

label_encoder = LabelEncoder()


y = label_encoder.fit_transform(
    y_text
)


print("\nClasses:")

for i, class_name in enumerate(
    label_encoder.classes_
):

    print(
        i,
        "=",
        class_name
    )


# ======================================================================
# STEP 24: TRAIN TEST SPLIT
# ======================================================================
# Split is performed BEFORE PSO.
# PSO sees only training data.
# The final test data therefore remains held out.
# ======================================================================

indices = np.arange(
    len(df)
)


try:

    train_indices, test_indices = train_test_split(

        indices,

        test_size=0.25,

        random_state=SEED,

        stratify=y
    )


except ValueError:

    print(
        "\nStratification not possible."
    )

    print(
        "Using normal train/test split."
    )


    train_indices, test_indices = train_test_split(

        indices,

        test_size=0.25,

        random_state=SEED
    )


X_train_original = (
    X_original
    .iloc[
        train_indices
    ]
    .reset_index(
        drop=True
    )
)


X_test_original = (
    X_original
    .iloc[
        test_indices
    ]
    .reset_index(
        drop=True
    )
)


y_train = y[
    train_indices
]


y_test = y[
    test_indices
]


print(
    "\nTraining samples:",
    len(y_train)
)


print(
    "Testing samples:",
    len(y_test)
)


# ======================================================================
# STEP 25: PSO PARAMETERS
# ======================================================================

NUMBER_OF_PARTICLES = 30

NUMBER_OF_ITERATIONS = 30


# Inertia weight
W = 0.7


# Cognitive coefficient
C1 = 1.5


# Social coefficient
C2 = 1.5


number_of_features = len(
    feature_columns
)


print("\n" + "=" * 80)

print(
    "STARTING BINARY PARTICLE SWARM OPTIMIZATION"
)

print("=" * 80)


print(
    "\nParticles:",
    NUMBER_OF_PARTICLES
)

print(
    "Iterations:",
    NUMBER_OF_ITERATIONS
)

print(
    "Features:",
    number_of_features
)


# ======================================================================
# STEP 26: SIGMOID FUNCTION
# ======================================================================
# Binary PSO uses sigmoid to convert velocity into probability.
# ======================================================================

def sigmoid(
    value
):

    value = np.clip(
        value,
        -20,
        20
    )

    return (
        1.0
        /
        (
            1.0
            +
            np.exp(
                -value
            )
        )
    )


# ======================================================================
# STEP 27: FITNESS FUNCTION
# ======================================================================
# Fitness = cross-validation accuracy - feature penalty.
#
# Higher fitness is better.
#
# The feature penalty encourages PSO to find a smaller subset rather
# than simply selecting every feature.
# ======================================================================

def calculate_fitness(
    particle
):

    selected_indices = np.where(
        particle == 1
    )[0]


    if len(
        selected_indices
    ) == 0:

        return 0.0


    selected_features = [

        feature_columns[i]

        for i in selected_indices
    ]


    X_selected = X_train_original[
        selected_features
    ]


    model = RandomForestClassifier(

        n_estimators=100,

        random_state=SEED,

        class_weight="balanced"
    )


    # --------------------------------------------------------------
    # SAFE CV FOR SMALL DATASET
    # --------------------------------------------------------------

    unique_classes, class_counts = np.unique(

        y_train,

        return_counts=True
    )


    minimum_class_count = (
        class_counts.min()
    )


    cv_folds = min(
        3,
        minimum_class_count
    )


    # --------------------------------------------------------------
    # CROSS VALIDATION
    # --------------------------------------------------------------

    if cv_folds >= 2:

        cv = StratifiedKFold(

            n_splits=cv_folds,

            shuffle=True,

            random_state=SEED
        )


        try:

            scores = cross_val_score(

                model,

                X_selected,

                y_train,

                cv=cv,

                scoring="accuracy"
            )


            accuracy = float(
                np.mean(
                    scores
                )
            )


        except Exception:

            model.fit(

                X_selected,

                y_train
            )


            prediction = model.predict(
                X_selected
            )


            accuracy = accuracy_score(

                y_train,

                prediction
            )


    else:

        model.fit(

            X_selected,

            y_train
        )


        prediction = model.predict(
            X_selected
        )


        accuracy = accuracy_score(

            y_train,

            prediction
        )


    # --------------------------------------------------------------
    # FEATURE PENALTY
    # --------------------------------------------------------------

    feature_ratio = (

        len(
            selected_features
        )

        /

        number_of_features
    )


    penalty = (
        0.02
        *
        feature_ratio
    )


    fitness = (
        accuracy
        -
        penalty
    )


    return float(
        fitness
    )


# ======================================================================
# STEP 28: INITIALIZE PARTICLES
# ======================================================================

positions = np.random.randint(

    0,
    2,

    size=(
        NUMBER_OF_PARTICLES,
        number_of_features
    )
)


# Ensure every particle has at least one feature

for i in range(
    NUMBER_OF_PARTICLES
):

    if np.sum(
        positions[i]
    ) == 0:

        random_feature = np.random.randint(
            0,
            number_of_features
        )

        positions[
            i,
            random_feature
        ] = 1


# ======================================================================
# STEP 29: INITIALIZE VELOCITIES
# ======================================================================

velocities = np.random.uniform(

    -1,

    1,

    size=(
        NUMBER_OF_PARTICLES,
        number_of_features
    )
)


# ======================================================================
# STEP 30: PERSONAL BESTS
# ======================================================================

personal_best_positions = (
    positions.copy()
)


personal_best_fitness = np.array([

    calculate_fitness(
        particle
    )

    for particle in positions
])


# ======================================================================
# STEP 31: GLOBAL BEST
# ======================================================================

global_best_index = np.argmax(
    personal_best_fitness
)


global_best_position = (
    personal_best_positions[
        global_best_index
    ].copy()
)


global_best_fitness = float(

    personal_best_fitness[
        global_best_index
    ]
)


print(
    "\nInitial Global Best Fitness:",
    round(
        global_best_fitness,
        4
    )
)


# ======================================================================
# STEP 32: PSO HISTORY
# ======================================================================

pso_history = []


# ======================================================================
# STEP 33: PSO OPTIMIZATION LOOP
# ======================================================================

for iteration in range(
    NUMBER_OF_ITERATIONS
):

    print(
        "\nIteration",
        iteration + 1,
        "/",
        NUMBER_OF_ITERATIONS
    )


    for particle_index in range(
        NUMBER_OF_PARTICLES
    ):


        # ----------------------------------------------------------
        # RANDOM COMPONENTS
        # ----------------------------------------------------------

        r1 = np.random.random(
            number_of_features
        )


        r2 = np.random.random(
            number_of_features
        )


        # ----------------------------------------------------------
        # VELOCITY UPDATE
        # ----------------------------------------------------------

        cognitive_component = (

            C1
            *
            r1
            *
            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )
        )


        social_component = (

            C2
            *
            r2
            *
            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )
        )


        velocities[
            particle_index
        ] = (

            W
            *
            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )


        # ----------------------------------------------------------
        # LIMIT VELOCITY
        # ----------------------------------------------------------

        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -6,

            6
        )


        # ----------------------------------------------------------
        # BINARY POSITION UPDATE
        # ----------------------------------------------------------

        probabilities = sigmoid(

            velocities[
                particle_index
            ]
        )


        random_values = np.random.random(
            number_of_features
        )


        new_position = (

            random_values
            <
            probabilities

        ).astype(
            int
        )


        # ----------------------------------------------------------
        # MAKE SURE AT LEAST ONE FEATURE IS SELECTED
        # ----------------------------------------------------------

        if np.sum(
            new_position
        ) == 0:

            random_feature = np.random.randint(

                0,

                number_of_features
            )


            new_position[
                random_feature
            ] = 1


        positions[
            particle_index
        ] = new_position


        # ----------------------------------------------------------
        # CALCULATE NEW FITNESS
        # ----------------------------------------------------------

        current_fitness = calculate_fitness(

            positions[
                particle_index
            ]
        )


        # ----------------------------------------------------------
        # UPDATE PERSONAL BEST
        # ----------------------------------------------------------

        if (
            current_fitness
            >
            personal_best_fitness[
                particle_index
            ]
        ):

            personal_best_fitness[
                particle_index
            ] = current_fitness


            personal_best_positions[
                particle_index
            ] = positions[
                particle_index
            ].copy()


        # ----------------------------------------------------------
        # UPDATE GLOBAL BEST
        # ----------------------------------------------------------

        if (
            current_fitness
            >
            global_best_fitness
        ):

            global_best_fitness = float(
                current_fitness
            )


            global_best_position = positions[
                particle_index
            ].copy()


    # --------------------------------------------------------------
    # SAVE HISTORY
    # --------------------------------------------------------------

    pso_history.append(
        global_best_fitness
    )


    selected_count = int(
        np.sum(
            global_best_position
        )
    )


    print(
        "Global Best Fitness:",
        round(
            global_best_fitness,
            4
        )
    )


    print(
        "Selected Features:",
        selected_count,
        "/",
        number_of_features
    )


# ======================================================================
# STEP 34: PSO COMPLETED
# ======================================================================

print("\n" + "=" * 80)

print(
    "PSO FEATURE SELECTION COMPLETED"
)

print("=" * 80)


selected_indices = np.where(

    global_best_position == 1

)[0]


selected_features = [

    feature_columns[i]

    for i in selected_indices
]


if len(
    selected_features
) == 0:

    raise ValueError(
        "PSO did not select any features."
    )


print(
    "\nBest PSO Fitness:",
    round(
        global_best_fitness,
        4
    )
)


print(
    "\nOriginal Feature Count:",
    len(
        feature_columns
    )
)


print(
    "PSO Selected Feature Count:",
    len(
        selected_features
    )
)


print("\nSelected Features:")


for feature in selected_features:

    print(
        "✓",
        feature
    )


# ======================================================================
# STEP 35: SAVE PSO FEATURE SELECTION CSV
# ======================================================================

feature_selection_df = pd.DataFrame({

    "Feature":
        feature_columns,

    "Selected_By_PSO": [

        "Yes"
        if value == 1
        else "No"

        for value in global_best_position
    ],

    "Binary_Value":
        global_best_position.astype(
            int
        )
})


feature_selection_df.to_csv(

    PSO_FEATURE_SELECTION_CSV,

    index=False
)


print(
    "\nSaved:",
    PSO_FEATURE_SELECTION_CSV
)


# ======================================================================
# STEP 36: PSO OPTIMIZATION GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    range(
        1,
        NUMBER_OF_ITERATIONS + 1
    ),

    pso_history,

    marker="o"
)


plt.xlabel(
    "PSO Iteration"
)


plt.ylabel(
    "Global Best Fitness"
)


plt.title(
    "PSO Feature Selection Optimization"
)


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_FEATURE_SELECTION_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_FEATURE_SELECTION_GRAPH
)


# ======================================================================
# STEP 37: CREATE SELECTED DATA
# ======================================================================

X_selected = df[
    selected_features
].copy()


X_train_selected = (
    X_selected
    .iloc[
        train_indices
    ]
    .reset_index(
        drop=True
    )
)


X_test_selected = (
    X_selected
    .iloc[
        test_indices
    ]
    .reset_index(
        drop=True
    )
)


# ======================================================================
# STEP 38: SCALE SELECTED FEATURES
# ======================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train_selected
)


X_test_scaled = scaler.transform(
    X_test_selected
)


X_full_scaled = scaler.transform(
    X_selected
)


# ======================================================================
# STEP 39: RANDOM FOREST USING PSO FEATURES
# ======================================================================

print("\n" + "=" * 80)

print(
    "TRAINING RANDOM FOREST USING PSO FEATURES"
)

print("=" * 80)


rf_model = RandomForestClassifier(

    n_estimators=300,

    random_state=SEED,

    class_weight="balanced"
)


rf_model.fit(

    X_train_scaled,

    y_train
)


rf_prediction = rf_model.predict(
    X_test_scaled
)


rf_probability = rf_model.predict_proba(
    X_test_scaled
)


# ======================================================================
# STEP 40: RF METRICS
# ======================================================================

rf_accuracy = accuracy_score(

    y_test,

    rf_prediction
)


rf_precision = precision_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0
)


rf_recall = recall_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0
)


rf_f1 = f1_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0
)


print(
    f"\nPSO Random Forest Accuracy: "
    f"{rf_accuracy * 100:.2f}%"
)


print("\nPSO Random Forest Classification Report:")


print(

    classification_report(

        y_test,

        rf_prediction,

        zero_division=0
    )
)


# ======================================================================
# STEP 41: BUILD PSO NEURAL NETWORK
# ======================================================================

print("\n" + "=" * 80)

print(
    "TRAINING NEURAL NETWORK USING PSO FEATURES"
)

print("=" * 80)


number_selected_features = (
    X_train_scaled.shape[1]
)


number_classes = len(
    label_encoder.classes_
)


nn_model = Sequential([

    Input(
        shape=(
            number_selected_features,
        )
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.10
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_classes,
        activation="softmax"
    )

])


# ======================================================================
# STEP 42: COMPILE NN
# ======================================================================

nn_model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


# ======================================================================
# STEP 43: EARLY STOPPING
# ======================================================================

early_stop = EarlyStopping(

    monitor="val_loss",

    patience=25,

    restore_best_weights=True
)


# ======================================================================
# STEP 44: TRAIN NN
# ======================================================================

history = nn_model.fit(

    X_train_scaled,

    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stop
    ],

    verbose=1
)


# ======================================================================
# STEP 45: NN PREDICTION
# ======================================================================

nn_probability = nn_model.predict(

    X_test_scaled,

    verbose=0
)


nn_prediction = np.argmax(

    nn_probability,

    axis=1
)


# ======================================================================
# STEP 46: NN METRICS
# ======================================================================

nn_accuracy = accuracy_score(

    y_test,

    nn_prediction
)


nn_precision = precision_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0
)


nn_recall = recall_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0
)


nn_f1 = f1_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0
)


print(
    f"\nPSO Neural Network Accuracy: "
    f"{nn_accuracy * 100:.2f}%"
)


print("\nPSO Neural Network Classification Report:")


print(

    classification_report(

        y_test,

        nn_prediction,

        zero_division=0
    )
)


# ======================================================================
# STEP 47: SAVE PSO H5 MODEL
# ======================================================================

nn_model.save(
    PSO_H5_PATH
)


print(
    "\nSaved:",
    PSO_H5_PATH
)


# ======================================================================
# STEP 48: SAVE PSO PKL
# ======================================================================

pkl_data = {

    "algorithm":
        "Binary Particle Swarm Optimization",

    "random_forest_model":
        rf_model,

    "scaler":
        scaler,

    "label_encoder":
        label_encoder,

    "all_features":
        feature_columns,

    "pso_selected_features":
        selected_features,

    "pso_best_position":
        global_best_position.tolist(),

    "pso_best_fitness":
        float(
            global_best_fitness
        ),

    "pso_history":
        [
            float(x)
            for x in pso_history
        ],

    "actual_columns":
        actual_columns,

    "normal_columns":
        normal_columns,

    "district_column":
        district_column,

    "classification_thresholds": {

        "deficient_below":
            -20,

        "normal_lower":
            -20,

        "normal_upper":
            20,

        "excess_above":
            20
    }
}


with open(

    PSO_PKL_PATH,

    "wb"

) as file:

    pickle.dump(

        pkl_data,

        file
    )


print(
    "Saved:",
    PSO_PKL_PATH
)


# ======================================================================
# STEP 49: SAVE PSO YAML
# ======================================================================

yaml_data = {

    "project": {

        "name":
            "AI-Based District Rainfall Deviation and Anomaly Analysis",

        "dataset":
            "Rainfalldistrict_2.csv",

        "optimization":
            "Binary Particle Swarm Optimization"
    },

    "pso_parameters": {

        "particles":
            NUMBER_OF_PARTICLES,

        "iterations":
            NUMBER_OF_ITERATIONS,

        "inertia_weight":
            W,

        "cognitive_coefficient":
            C1,

        "social_coefficient":
            C2
    },

    "feature_selection": {

        "original_feature_count":
            int(
                len(
                    feature_columns
                )
            ),

        "selected_feature_count":
            int(
                len(
                    selected_features
                )
            ),

        "best_fitness":
            float(
                global_best_fitness
            ),

        "selected_features": [

            str(x)

            for x in selected_features
        ]
    },

    "random_forest": {

        "accuracy":
            float(
                rf_accuracy
            ),

        "precision":
            float(
                rf_precision
            ),

        "recall":
            float(
                rf_recall
            ),

        "f1_score":
            float(
                rf_f1
            )
    },

    "neural_network": {

        "accuracy":
            float(
                nn_accuracy
            ),

        "precision":
            float(
                nn_precision
            ),

        "recall":
            float(
                nn_recall
            ),

        "f1_score":
            float(
                nn_f1
            )
    }
}


with open(

    PSO_YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_data,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "Saved:",
    PSO_YAML_PATH
)


# ======================================================================
# STEP 50: SAVE PSO JSON
# ======================================================================

json_data = {

    "project":
        "AI-Based District Rainfall Deviation and Anomaly Analysis",

    "dataset":
        "Rainfalldistrict_2.csv",

    "optimization_algorithm":
        "Binary Particle Swarm Optimization",

    "pso_parameters": {

        "particles":
            NUMBER_OF_PARTICLES,

        "iterations":
            NUMBER_OF_ITERATIONS,

        "inertia_weight":
            W,

        "cognitive_coefficient":
            C1,

        "social_coefficient":
            C2
    },

    "original_features": [

        str(x)

        for x in feature_columns
    ],

    "selected_features": [

        str(x)

        for x in selected_features
    ],

    "best_particle": [

        int(x)

        for x in global_best_position
    ],

    "best_fitness":
        float(
            global_best_fitness
        ),

    "random_forest_metrics": {

        "accuracy":
            float(
                rf_accuracy
            ),

        "precision":
            float(
                rf_precision
            ),

        "recall":
            float(
                rf_recall
            ),

        "f1_score":
            float(
                rf_f1
            )
    },

    "neural_network_metrics": {

        "accuracy":
            float(
                nn_accuracy
            ),

        "precision":
            float(
                nn_precision
            ),

        "recall":
            float(
                nn_recall
            ),

        "f1_score":
            float(
                nn_f1
            )
    }
}


with open(

    PSO_JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_data,

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "Saved:",
    PSO_JSON_PATH
)


# ======================================================================
# STEP 51: PSO ACCURACY GRAPH
# ======================================================================

print(
    "\nCreating pso_accuracy_graph.png..."
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"
)


plt.plot(

    history.history[
        "val_accuracy"
    ],

    marker="s",

    markersize=3,

    label="Validation Accuracy"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.title(
    "PSO Neural Network Training and Validation Accuracy"
)


plt.ylim(
    0,
    1.05
)


plt.grid(
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    PSO_ACCURACY_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_ACCURACY_GRAPH
)


# ======================================================================
# STEP 52: PSO HEATMAP
# ======================================================================

print(
    "\nCreating pso_heatmap.png..."
)


cm = confusion_matrix(

    y_test,

    rf_prediction,

    labels=np.arange(
        len(
            label_encoder.classes_
        )
    )
)


plt.figure(
    figsize=(8, 6)
)


plt.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"
)


plt.title(
    "PSO Random Forest Confusion Matrix"
)


plt.colorbar()


tick_marks = np.arange(

    len(
        label_encoder.classes_
    )
)


plt.xticks(

    tick_marks,

    label_encoder.classes_,

    rotation=45
)


plt.yticks(

    tick_marks,

    label_encoder.classes_
)


threshold = (

    cm.max() / 2

    if cm.size > 0

    else 0
)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            horizontalalignment="center",

            verticalalignment="center",

            color=(

                "white"

                if cm[i, j] > threshold

                else "black"
            )
        )


plt.xlabel(
    "Predicted Category"
)


plt.ylabel(
    "Actual Category"
)


plt.tight_layout()


plt.savefig(

    PSO_HEATMAP_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_HEATMAP_GRAPH
)


# ======================================================================
# STEP 53: PSO MODEL COMPARISON
# ======================================================================

comparison_df = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score"
    ],

    "PSO Random Forest": [

        rf_accuracy,

        rf_precision,

        rf_recall,

        rf_f1
    ],

    "PSO Neural Network": [

        nn_accuracy,

        nn_precision,

        nn_recall,

        nn_f1
    ]
})


print("\nPSO Model Comparison:")

print(
    comparison_df
)


x = np.arange(
    len(
        comparison_df
    )
)


width = 0.35


plt.figure(
    figsize=(10, 6)
)


bars1 = plt.bar(

    x - width / 2,

    comparison_df[
        "PSO Random Forest"
    ],

    width,

    label="PSO Random Forest"
)


bars2 = plt.bar(

    x + width / 2,

    comparison_df[
        "PSO Neural Network"
    ],

    width,

    label="PSO Neural Network"
)


plt.xticks(

    x,

    comparison_df[
        "Metric"
    ]
)


plt.xlabel(
    "Evaluation Metric"
)


plt.ylabel(
    "Score"
)


plt.title(
    "PSO Optimized Model Comparison"
)


plt.ylim(
    0,
    1.10
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.legend()


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        height = (
            bar.get_height()
        )


        plt.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            height + 0.02,

            f"{height:.2f}",

            ha="center",

            va="bottom"
        )


plt.tight_layout()


plt.savefig(

    PSO_COMPARISON_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_COMPARISON_GRAPH
)


# ======================================================================
# STEP 54: CREATE PSO RESULT CSV
# ======================================================================

pso_result_df = df[

    [
        district_column,

        "Calculated_Total_Actual",

        "Calculated_Total_Normal",

        "Rainfall_Deviation_Percentage",

        "Rainfall_Category"
    ]

].copy()


pso_result_df.columns = [

    "District",

    "Actual_Rainfall",

    "Normal_Rainfall",

    "Deviation_Percentage",

    "Rainfall_Category"
]


pso_result_df[
    "Actual_Rainfall"
] = pso_result_df[
    "Actual_Rainfall"
].round(
    2
)


pso_result_df[
    "Normal_Rainfall"
] = pso_result_df[
    "Normal_Rainfall"
].round(
    2
)


pso_result_df[
    "Deviation_Percentage"
] = pso_result_df[
    "Deviation_Percentage"
].round(
    2
)


pso_result_df.to_csv(

    PSO_RESULT_CSV,

    index=False
)


print(
    "\nSaved:",
    PSO_RESULT_CSV
)


print("\nPSO Results:")

print(
    pso_result_df
)


# ======================================================================
# STEP 55: PSO RESULT GRAPH
# ======================================================================

result_plot = pso_result_df.sort_values(

    by="Deviation_Percentage"

).reset_index(
    drop=True
)


plt.figure(
    figsize=(14, 9)
)


plt.barh(

    result_plot[
        "District"
    ],

    result_plot[
        "Deviation_Percentage"
    ]
)


plt.axvline(
    x=0,
    linewidth=1
)


plt.axvline(

    x=-20,

    linestyle="--",

    linewidth=1,

    label="Deficient Threshold (-20%)"
)


plt.axvline(

    x=20,

    linestyle="--",

    linewidth=1,

    label="Excess Threshold (+20%)"
)


plt.xlabel(
    "Rainfall Deviation (%)"
)


plt.ylabel(
    "District"
)


plt.title(
    "PSO District-wise Rainfall Deviation Analysis"
)


plt.grid(
    axis="x",
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    PSO_RESULT_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_RESULT_GRAPH
)


# ======================================================================
# STEP 56: PREDICTIONS FOR COMPLETE DATASET
# ======================================================================

full_rf_prediction = rf_model.predict(
    X_full_scaled
)


full_rf_probability = rf_model.predict_proba(
    X_full_scaled
)


full_nn_probability = nn_model.predict(

    X_full_scaled,

    verbose=0
)


full_nn_prediction = np.argmax(

    full_nn_probability,

    axis=1
)


# ======================================================================
# STEP 57: DECODE PREDICTIONS
# ======================================================================

rf_prediction_names = (
    label_encoder.inverse_transform(
        full_rf_prediction
    )
)


nn_prediction_names = (
    label_encoder.inverse_transform(
        full_nn_prediction
    )
)


rf_confidence = np.max(

    full_rf_probability,

    axis=1

) * 100


nn_confidence = np.max(

    full_nn_probability,

    axis=1

) * 100


# ======================================================================
# STEP 58: PSO PREDICTION CSV
# ======================================================================

pso_prediction_df = pd.DataFrame({

    "District":

        df[
            district_column
        ].values,

    "Actual_Rainfall":

        df[
            "Calculated_Total_Actual"
        ].round(
            2
        ),

    "Normal_Rainfall":

        df[
            "Calculated_Total_Normal"
        ].round(
            2
        ),

    "Deviation_Percentage":

        df[
            "Rainfall_Deviation_Percentage"
        ].round(
            2
        ),

    "Actual_Category":

        df[
            "Rainfall_Category"
        ].values,

    "PSO_RF_Predicted_Category":

        rf_prediction_names,

    "PSO_RF_Confidence_Percentage":

        np.round(
            rf_confidence,
            2
        ),

    "PSO_NN_Predicted_Category":

        nn_prediction_names,

    "PSO_NN_Confidence_Percentage":

        np.round(
            nn_confidence,
            2
        )
})


pso_prediction_df.to_csv(

    PSO_PREDICTION_CSV,

    index=False
)


print(
    "\nSaved:",
    PSO_PREDICTION_CSV
)


print("\nPSO Predictions:")

print(
    pso_prediction_df
)


# ======================================================================
# STEP 59: PSO PREDICTION GRAPH
# ======================================================================

prediction_plot = (
    pso_prediction_df
    .sort_values(
        by="Deviation_Percentage"
    )
    .reset_index(
        drop=True
    )
)


x = np.arange(
    len(
        prediction_plot
    )
)


plt.figure(
    figsize=(16, 8)
)


plt.plot(

    x,

    prediction_plot[
        "PSO_RF_Confidence_Percentage"
    ],

    marker="o",

    label="PSO Random Forest Confidence"
)


plt.plot(

    x,

    prediction_plot[
        "PSO_NN_Confidence_Percentage"
    ],

    marker="s",

    label="PSO Neural Network Confidence"
)


plt.xticks(

    x,

    prediction_plot[
        "District"
    ],

    rotation=90
)


plt.xlabel(
    "District"
)


plt.ylabel(
    "Prediction Confidence (%)"
)


plt.title(
    "PSO District-wise Rainfall Prediction Confidence"
)


plt.ylim(
    0,
    105
)


plt.grid(
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    PSO_PREDICTION_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_PREDICTION_GRAPH
)


# ======================================================================
# STEP 60: ACTUAL VS NORMAL GRAPH
# ======================================================================

x = np.arange(
    len(
        pso_result_df
    )
)


width = 0.40


plt.figure(
    figsize=(16, 8)
)


plt.bar(

    x - width / 2,

    pso_result_df[
        "Actual_Rainfall"
    ],

    width,

    label="Actual Rainfall"
)


plt.bar(

    x + width / 2,

    pso_result_df[
        "Normal_Rainfall"
    ],

    width,

    label="Normal Rainfall"
)


plt.xticks(

    x,

    pso_result_df[
        "District"
    ],

    rotation=90
)


plt.xlabel(
    "District"
)


plt.ylabel(
    "Rainfall"
)


plt.title(
    "PSO Actual vs Normal Rainfall by District"
)


plt.grid(
    axis="y",
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    PSO_ACTUAL_NORMAL_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_ACTUAL_NORMAL_GRAPH
)


# ======================================================================
# STEP 61: CATEGORY DISTRIBUTION
# ======================================================================

category_counts = pso_result_df[

    "Rainfall_Category"

].value_counts()


plt.figure(
    figsize=(8, 6)
)


bars = plt.bar(

    category_counts.index,

    category_counts.values
)


plt.xlabel(
    "Rainfall Category"
)


plt.ylabel(
    "Number of Districts"
)


plt.title(
    "PSO Rainfall Category Distribution"
)


plt.grid(
    axis="y",
    alpha=0.3
)


for bar in bars:

    height = (
        bar.get_height()
    )


    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height + 0.1,

        str(
            int(
                height
            )
        ),

        ha="center"
    )


plt.tight_layout()


plt.savefig(

    PSO_CATEGORY_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_CATEGORY_GRAPH
)


# ======================================================================
# STEP 62: FEATURE IMPORTANCE AFTER PSO
# ======================================================================

importance_df = pd.DataFrame({

    "Feature":
        selected_features,

    "Importance":
        rf_model.feature_importances_
})


importance_df = importance_df.sort_values(

    by="Importance",

    ascending=True
)


plt.figure(
    figsize=(10, 7)
)


plt.barh(

    importance_df[
        "Feature"
    ],

    importance_df[
        "Importance"
    ]
)


plt.xlabel(
    "Feature Importance"
)


plt.ylabel(
    "PSO Selected Feature"
)


plt.title(
    "Random Forest Feature Importance After PSO Selection"
)


plt.grid(
    axis="x",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_FEATURE_IMPORTANCE_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    PSO_FEATURE_IMPORTANCE_GRAPH
)


# ======================================================================
# STEP 63: TEST SAVED PSO PKL MODEL
# ======================================================================

print("\n" + "=" * 80)

print(
    "TESTING SAVED PSO PKL MODEL"
)

print("=" * 80)


with open(

    PSO_PKL_PATH,

    "rb"

) as file:

    loaded_data = pickle.load(
        file
    )


loaded_rf_model = loaded_data[
    "random_forest_model"
]


loaded_scaler = loaded_data[
    "scaler"
]


loaded_label_encoder = loaded_data[
    "label_encoder"
]


loaded_features = loaded_data[
    "pso_selected_features"
]


sample = df[
    loaded_features
].iloc[
    [0]
]


sample_scaled = loaded_scaler.transform(
    sample
)


sample_prediction = loaded_rf_model.predict(
    sample_scaled
)


sample_category = (
    loaded_label_encoder.inverse_transform(
        sample_prediction
    )[0]
)


print("\nSample District:")

print(
    df.iloc[0][
        district_column
    ]
)


print("\nPSO PKL Model Prediction:")

print(
    sample_category
)


# ======================================================================
# STEP 64: TEST SAVED PSO H5 MODEL
# ======================================================================

print("\n" + "=" * 80)

print(
    "TESTING SAVED PSO H5 MODEL"
)

print("=" * 80)


loaded_nn_model = tf.keras.models.load_model(
    PSO_H5_PATH
)


sample_nn_probability = loaded_nn_model.predict(

    sample_scaled,

    verbose=0
)


sample_nn_prediction = np.argmax(

    sample_nn_probability,

    axis=1
)


sample_nn_category = (
    loaded_label_encoder.inverse_transform(
        sample_nn_prediction
    )[0]
)


print("\nPSO H5 Model Prediction:")

print(
    sample_nn_category
)


# ======================================================================
# STEP 65: FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 80)

print(
    "PSO PROJECT COMPLETED SUCCESSFULLY"
)

print("=" * 80)


print(
    "\nOriginal Features:",
    len(
        feature_columns
    )
)


print(
    "PSO Selected Features:",
    len(
        selected_features
    )
)


print(
    "PSO Best Fitness:",
    round(
        global_best_fitness,
        4
    )
)


print(
    f"\nPSO Random Forest Accuracy: "
    f"{rf_accuracy * 100:.2f}%"
)


print(
    f"PSO Neural Network Accuracy: "
    f"{nn_accuracy * 100:.2f}%"
)


print("\nPSO Selected Features:")


for feature in selected_features:

    print(
        " -",
        feature
    )


# ======================================================================
# STEP 66: GENERATED FILES
# ======================================================================

generated_files = [

    PSO_H5_PATH,

    PSO_PKL_PATH,

    PSO_YAML_PATH,

    PSO_JSON_PATH,

    PSO_ACCURACY_GRAPH,

    PSO_HEATMAP_GRAPH,

    PSO_COMPARISON_GRAPH,

    PSO_RESULT_CSV,

    PSO_RESULT_GRAPH,

    PSO_PREDICTION_CSV,

    PSO_PREDICTION_GRAPH,

    PSO_FEATURE_SELECTION_CSV,

    PSO_FEATURE_SELECTION_GRAPH,

    PSO_FEATURE_IMPORTANCE_GRAPH,

    PSO_ACTUAL_NORMAL_GRAPH,

    PSO_CATEGORY_GRAPH
]


print("\nGenerated PSO Files:")


for number, path in enumerate(

    generated_files,

    start=1
):

    print(
        f"{number}. {path}"
    )


print("\n" + "=" * 80)

print(
    "ALL PSO OUTPUTS SAVED SUCCESSFULLY"
)

print("=" * 80)
```
