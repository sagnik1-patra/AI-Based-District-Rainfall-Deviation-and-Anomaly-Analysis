```python
import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping


# ======================================================================
# STEP 1: PATHS
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District Rainfall Deviation and Anomaly Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "Rainfalldistrict_2.csv"
)

ACCURACY_GRAPH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)

HEATMAP_GRAPH = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)

COMPARISON_GRAPH = os.path.join(
    PROJECT_DIR,
    "model_comparison_graph.png"
)

RESULT_CSV = os.path.join(
    PROJECT_DIR,
    "result.csv"
)

RESULT_GRAPH = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)

PREDICTION_CSV = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)

os.makedirs(PROJECT_DIR, exist_ok=True)


# ======================================================================
# STEP 2: LOAD DATASET
# ======================================================================

print("=" * 75)
print("AI-BASED DISTRICT RAINFALL DEVIATION AND ANOMALY ANALYSIS")
print("=" * 75)

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"Dataset not found:\n{CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)


# ======================================================================
# STEP 3: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)

df = df.reset_index(drop=True)

print("\nColumns:")

for col in df.columns:
    print(col)


# ======================================================================
# STEP 4: FIND DISTRICT COLUMN
# ======================================================================

district_column = None

for col in df.columns:

    if "district" in col.lower():
        district_column = col
        break


if district_column is None:
    district_column = df.columns[0]


print("\nDistrict column:")
print(district_column)


# ======================================================================
# STEP 5: CLEAN DISTRICT NAMES
# ======================================================================

df[district_column] = (
    df[district_column]
    .astype(str)
    .str.strip()
)

# Remove rows where district name is missing
df = df[
    ~df[district_column]
    .str.lower()
    .isin(["nan", "none", ""])
].copy()

df.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# STEP 6: CONVERT POSSIBLE RAINFALL COLUMNS TO NUMERIC
# ======================================================================

for col in df.columns:

    if col == district_column:
        continue

    cleaned = (
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    # Convert common missing-value markers
    cleaned = cleaned.replace(
        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None"
        ],
        np.nan
    )

    numeric_version = pd.to_numeric(
        cleaned,
        errors="coerce"
    )

    # Only convert when column actually contains numeric information
    if numeric_version.notna().sum() > 0:
        df[col] = numeric_version


# ======================================================================
# STEP 7: FIND NUMERIC COLUMNS
# ======================================================================

numeric_columns = df.select_dtypes(
    include=[np.number]
).columns.tolist()

print("\nNumeric columns:")

for col in numeric_columns:
    print(col)


if len(numeric_columns) < 2:
    raise ValueError(
        "Not enough numerical rainfall columns detected."
    )


# ======================================================================
# STEP 8: HANDLE MISSING VALUES
# ======================================================================

for col in numeric_columns:

    median_value = df[col].median()

    if pd.isna(median_value):
        median_value = 0

    df[col] = df[col].fillna(
        median_value
    )


# ======================================================================
# STEP 9: IDENTIFY ACTUAL AND NORMAL COLUMNS
# ======================================================================

actual_columns = []
normal_columns = []

for col in numeric_columns:

    lower = col.lower()

    if "actual" in lower:
        actual_columns.append(col)

    elif "normal" in lower:
        normal_columns.append(col)


# ======================================================================
# STEP 10: FALLBACK FOR ALTERNATING COLUMNS
# ======================================================================

if (
    len(actual_columns) == 0
    or len(normal_columns) == 0
):

    print(
        "\nActual/Normal names were not detected."
    )

    print(
        "Using alternating numeric columns as Actual/Normal pairs."
    )

    actual_columns = numeric_columns[0::2]

    normal_columns = numeric_columns[1::2]


pair_count = min(
    len(actual_columns),
    len(normal_columns)
)

actual_columns = actual_columns[
    :pair_count
]

normal_columns = normal_columns[
    :pair_count
]


if pair_count == 0:
    raise ValueError(
        "Could not create Actual/Normal rainfall pairs."
    )


print("\nActual rainfall columns:")
print(actual_columns)

print("\nNormal rainfall columns:")
print(normal_columns)


# ======================================================================
# STEP 11: CALCULATE TOTAL ACTUAL RAINFALL
# ======================================================================

df["Calculated_Total_Actual"] = df[
    actual_columns
].sum(axis=1)


# ======================================================================
# STEP 12: CALCULATE TOTAL NORMAL RAINFALL
# ======================================================================

df["Calculated_Total_Normal"] = df[
    normal_columns
].sum(axis=1)


# ======================================================================
# STEP 13: CALCULATE DEVIATION
# ======================================================================

df["Rainfall_Deviation_Percentage"] = np.where(

    df["Calculated_Total_Normal"] != 0,

    (
        (
            df["Calculated_Total_Actual"]
            -
            df["Calculated_Total_Normal"]
        )
        /
        df["Calculated_Total_Normal"]
    ) * 100,

    0
)


# ======================================================================
# STEP 14: CREATE RAINFALL CLASS
# ======================================================================

def rainfall_category(value):

    if value < -20:
        return "Deficient"

    elif value <= 20:
        return "Normal"

    else:
        return "Excess"


df["Rainfall_Category"] = df[
    "Rainfall_Deviation_Percentage"
].apply(
    rainfall_category
)


print("\nRainfall category distribution:")

print(
    df["Rainfall_Category"]
    .value_counts()
)


# ======================================================================
# STEP 15: MACHINE LEARNING FEATURES
# ======================================================================

feature_columns = list(
    dict.fromkeys(
        actual_columns
        +
        normal_columns
    )
)

X = df[
    feature_columns
].copy()

y_text = df[
    "Rainfall_Category"
].copy()


# ======================================================================
# STEP 16: LABEL ENCODING
# ======================================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    y_text
)

print("\nClasses:")

for i, class_name in enumerate(
    label_encoder.classes_
):
    print(
        i,
        "=",
        class_name
    )


# ======================================================================
# STEP 17: FEATURE SCALING
# ======================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    X
)


# ======================================================================
# STEP 18: TRAIN TEST SPLIT
# ======================================================================

try:

    X_train, X_test, y_train, y_test = train_test_split(

        X_scaled,
        y,

        test_size=0.25,

        random_state=42,

        stratify=y
    )

except ValueError:

    print(
        "\nStratified split not possible."
    )

    X_train, X_test, y_train, y_test = train_test_split(

        X_scaled,
        y,

        test_size=0.25,

        random_state=42
    )


print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# ======================================================================
# STEP 19: RANDOM FOREST
# ======================================================================

print("\n" + "=" * 75)
print("RANDOM FOREST")
print("=" * 75)

rf_model = RandomForestClassifier(

    n_estimators=300,

    random_state=42,

    class_weight="balanced"
)

rf_model.fit(
    X_train,
    y_train
)

rf_prediction = rf_model.predict(
    X_test
)

rf_accuracy = accuracy_score(
    y_test,
    rf_prediction
)

rf_precision = precision_score(
    y_test,
    rf_prediction,
    average="weighted",
    zero_division=0
)

rf_recall = recall_score(
    y_test,
    rf_prediction,
    average="weighted",
    zero_division=0
)

rf_f1 = f1_score(
    y_test,
    rf_prediction,
    average="weighted",
    zero_division=0
)

print(
    f"\nRandom Forest Accuracy: "
    f"{rf_accuracy * 100:.2f}%"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        rf_prediction,
        zero_division=0
    )
)


# ======================================================================
# STEP 20: NEURAL NETWORK
# ======================================================================

print("\n" + "=" * 75)
print("NEURAL NETWORK")
print("=" * 75)

number_of_features = X_train.shape[1]

number_of_classes = len(
    label_encoder.classes_
)

nn_model = Sequential([

    Input(
        shape=(number_of_features,)
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(0.20),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(0.10),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )
])


nn_model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


early_stop = EarlyStopping(

    monitor="val_loss",

    patience=25,

    restore_best_weights=True
)


history = nn_model.fit(

    X_train,
    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stop
    ],

    verbose=1
)


# ======================================================================
# STEP 21: NEURAL NETWORK PREDICTIONS
# ======================================================================

nn_probability = nn_model.predict(
    X_test,
    verbose=0
)

nn_prediction = np.argmax(
    nn_probability,
    axis=1
)

nn_accuracy = accuracy_score(
    y_test,
    nn_prediction
)

nn_precision = precision_score(
    y_test,
    nn_prediction,
    average="weighted",
    zero_division=0
)

nn_recall = recall_score(
    y_test,
    nn_prediction,
    average="weighted",
    zero_division=0
)

nn_f1 = f1_score(
    y_test,
    nn_prediction,
    average="weighted",
    zero_division=0
)


print(
    f"\nNeural Network Accuracy: "
    f"{nn_accuracy * 100:.2f}%"
)


print("\nClassification Report:")

print(
    classification_report(
        y_test,
        nn_prediction,
        zero_division=0
    )
)


# ======================================================================
# STEP 22: ACCURACY GRAPH
# ======================================================================

print("\nCreating accuracy graph...")

plt.figure(
    figsize=(10, 6)
)

plt.plot(
    history.history["accuracy"],
    marker="o",
    markersize=3,
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    markersize=3,
    label="Validation Accuracy"
)

plt.title(
    "Neural Network Training and Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.ylim(
    0,
    1.05
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.savefig(
    ACCURACY_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

print(
    "Saved:",
    ACCURACY_GRAPH
)


# ======================================================================
# STEP 23: CONFUSION MATRIX HEATMAP
# ======================================================================

print("\nCreating heatmap...")

cm = confusion_matrix(
    y_test,
    rf_prediction,
    labels=np.arange(
        len(label_encoder.classes_)
    )
)

plt.figure(
    figsize=(8, 6)
)

plt.imshow(
    cm,
    interpolation="nearest",
    cmap="Blues"
)

plt.title(
    "Random Forest Confusion Matrix"
)

plt.colorbar()

tick_marks = np.arange(
    len(label_encoder.classes_)
)

plt.xticks(
    tick_marks,
    label_encoder.classes_,
    rotation=45
)

plt.yticks(
    tick_marks,
    label_encoder.classes_
)

threshold = (
    cm.max() / 2
    if cm.size > 0
    else 0
)

for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(
            j,
            i,
            str(cm[i, j]),
            horizontalalignment="center",
            verticalalignment="center",
            color=(
                "white"
                if cm[i, j] > threshold
                else "black"
            )
        )

plt.xlabel(
    "Predicted Category"
)

plt.ylabel(
    "Actual Category"
)

plt.tight_layout()

plt.savefig(
    HEATMAP_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

print(
    "Saved:",
    HEATMAP_GRAPH
)


# ======================================================================
# STEP 24: MODEL COMPARISON GRAPH
# ======================================================================

print("\nCreating model comparison graph...")

comparison_df = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],

    "Random Forest": [
        rf_accuracy,
        rf_precision,
        rf_recall,
        rf_f1
    ],

    "Neural Network": [
        nn_accuracy,
        nn_precision,
        nn_recall,
        nn_f1
    ]
})


print("\nModel Comparison:")
print(comparison_df)


x = np.arange(
    len(comparison_df)
)

width = 0.35


plt.figure(
    figsize=(10, 6)
)

bars1 = plt.bar(
    x - width / 2,
    comparison_df["Random Forest"],
    width,
    label="Random Forest"
)

bars2 = plt.bar(
    x + width / 2,
    comparison_df["Neural Network"],
    width,
    label="Neural Network"
)

plt.xticks(
    x,
    comparison_df["Metric"]
)

plt.ylabel(
    "Score"
)

plt.xlabel(
    "Evaluation Metric"
)

plt.title(
    "Random Forest vs Neural Network"
)

plt.ylim(
    0,
    1.1
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        height = bar.get_height()

        plt.text(
            bar.get_x()
            +
            bar.get_width() / 2,

            height + 0.02,

            f"{height:.2f}",

            ha="center",
            va="bottom"
        )


plt.tight_layout()

plt.savefig(
    COMPARISON_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

print(
    "Saved:",
    COMPARISON_GRAPH
)


# ======================================================================
# STEP 25: CREATE RESULT CSV
# ======================================================================

print("\nCreating result.csv...")

result_df = df[
    [
        district_column,
        "Calculated_Total_Actual",
        "Calculated_Total_Normal",
        "Rainfall_Deviation_Percentage",
        "Rainfall_Category"
    ]
].copy()


result_df.columns = [

    "District",

    "Actual_Rainfall",

    "Normal_Rainfall",

    "Deviation_Percentage",

    "Rainfall_Category"
]


result_df["Actual_Rainfall"] = (
    result_df["Actual_Rainfall"]
    .round(2)
)

result_df["Normal_Rainfall"] = (
    result_df["Normal_Rainfall"]
    .round(2)
)

result_df["Deviation_Percentage"] = (
    result_df["Deviation_Percentage"]
    .round(2)
)


result_df.to_csv(
    RESULT_CSV,
    index=False
)


print("\nResult CSV:")
print(result_df)

print(
    "\nSaved:",
    RESULT_CSV
)


# ======================================================================
# STEP 26: RESULT GRAPH
# ======================================================================

print("\nCreating result graph...")

result_plot_df = result_df.sort_values(
    by="Deviation_Percentage"
).reset_index(
    drop=True
)


plt.figure(
    figsize=(14, 8)
)

bars = plt.barh(
    result_plot_df["District"],
    result_plot_df["Deviation_Percentage"]
)

plt.axvline(
    x=0,
    linewidth=1
)

plt.axvline(
    x=-20,
    linestyle="--",
    linewidth=1,
    label="Deficient threshold (-20%)"
)

plt.axvline(
    x=20,
    linestyle="--",
    linewidth=1,
    label="Excess threshold (+20%)"
)

plt.title(
    "District-wise Rainfall Deviation"
)

plt.xlabel(
    "Rainfall Deviation (%)"
)

plt.ylabel(
    "District"
)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.savefig(
    RESULT_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

print(
    "Saved:",
    RESULT_GRAPH
)


# ======================================================================
# STEP 27: MAKE PREDICTIONS FOR COMPLETE DATASET
# ======================================================================

print("\nGenerating predictions...")

full_rf_prediction = rf_model.predict(
    X_scaled
)

full_rf_probability = rf_model.predict_proba(
    X_scaled
)

full_nn_probability = nn_model.predict(
    X_scaled,
    verbose=0
)

full_nn_prediction = np.argmax(
    full_nn_probability,
    axis=1
)


# ======================================================================
# STEP 28: CONVERT PREDICTIONS BACK TO CATEGORY NAMES
# ======================================================================

rf_prediction_names = label_encoder.inverse_transform(
    full_rf_prediction
)

nn_prediction_names = label_encoder.inverse_transform(
    full_nn_prediction
)


rf_confidence = np.max(
    full_rf_probability,
    axis=1
) * 100

nn_confidence = np.max(
    full_nn_probability,
    axis=1
) * 100


# ======================================================================
# STEP 29: CREATE PREDICTION CSV
# ======================================================================

prediction_df = pd.DataFrame({

    "District":
        df[district_column].values,

    "Actual_Rainfall":
        df["Calculated_Total_Actual"].round(2),

    "Normal_Rainfall":
        df["Calculated_Total_Normal"].round(2),

    "Deviation_Percentage":
        df[
            "Rainfall_Deviation_Percentage"
        ].round(2),

    "Actual_Category":
        df[
            "Rainfall_Category"
        ].values,

    "RF_Predicted_Category":
        rf_prediction_names,

    "RF_Confidence_Percentage":
        np.round(
            rf_confidence,
            2
        ),

    "NN_Predicted_Category":
        nn_prediction_names,

    "NN_Confidence_Percentage":
        np.round(
            nn_confidence,
            2
        )
})


prediction_df.to_csv(
    PREDICTION_CSV,
    index=False
)


print("\nPrediction CSV:")
print(prediction_df)

print(
    "\nSaved:",
    PREDICTION_CSV
)


# ======================================================================
# STEP 30: PREDICTION GRAPH
# ======================================================================

print("\nCreating prediction graph...")

prediction_plot_df = (
    prediction_df
    .sort_values(
        by="Deviation_Percentage"
    )
    .reset_index(
        drop=True
    )
)


x = np.arange(
    len(prediction_plot_df)
)


plt.figure(
    figsize=(16, 8)
)

plt.plot(
    x,
    prediction_plot_df[
        "RF_Confidence_Percentage"
    ],
    marker="o",
    label="Random Forest Confidence"
)

plt.plot(
    x,
    prediction_plot_df[
        "NN_Confidence_Percentage"
    ],
    marker="s",
    label="Neural Network Confidence"
)

plt.xticks(
    x,
    prediction_plot_df[
        "District"
    ],
    rotation=90
)

plt.ylabel(
    "Prediction Confidence (%)"
)

plt.xlabel(
    "District"
)

plt.title(
    "District-wise Rainfall Prediction Confidence"
)

plt.ylim(
    0,
    105
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.savefig(
    PREDICTION_GRAPH,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()

print(
    "Saved:",
    PREDICTION_GRAPH
)


# ======================================================================
# STEP 31: ACTUAL VS NORMAL RAINFALL GRAPH
# Extra useful project visualization
# ======================================================================

actual_normal_graph = os.path.join(
    PROJECT_DIR,
    "actual_vs_normal_rainfall.png"
)

plt.figure(
    figsize=(16, 8)
)

x = np.arange(
    len(result_df)
)

width = 0.40

plt.bar(
    x - width / 2,
    result_df[
        "Actual_Rainfall"
    ],
    width,
    label="Actual Rainfall"
)

plt.bar(
    x + width / 2,
    result_df[
        "Normal_Rainfall"
    ],
    width,
    label="Normal Rainfall"
)

plt.xticks(
    x,
    result_df["District"],
    rotation=90
)

plt.xlabel(
    "District"
)

plt.ylabel(
    "Rainfall"
)

plt.title(
    "Actual vs Normal Rainfall by District"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    actual_normal_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ======================================================================
# STEP 32: CATEGORY DISTRIBUTION GRAPH
# Extra useful visualization
# ======================================================================

category_graph = os.path.join(
    PROJECT_DIR,
    "rainfall_category_distribution.png"
)

category_counts = (
    result_df[
        "Rainfall_Category"
    ]
    .value_counts()
)


plt.figure(
    figsize=(8, 6)
)

bars = plt.bar(
    category_counts.index,
    category_counts.values
)

plt.xlabel(
    "Rainfall Category"
)

plt.ylabel(
    "Number of Districts"
)

plt.title(
    "Rainfall Category Distribution"
)

plt.grid(
    axis="y",
    alpha=0.3
)


for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x()
        +
        bar.get_width() / 2,

        height + 0.1,

        str(int(height)),

        ha="center"
    )


plt.tight_layout()

plt.savefig(
    category_graph,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

plt.close()


# ======================================================================
# STEP 33: FINAL OUTPUT SUMMARY
# ======================================================================

print("\n" + "=" * 75)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 75)

print(
    f"\nRandom Forest Accuracy: "
    f"{rf_accuracy * 100:.2f}%"
)

print(
    f"Neural Network Accuracy: "
    f"{nn_accuracy * 100:.2f}%"
)


print("\nGenerated Files:")

generated_files = [

    ACCURACY_GRAPH,

    HEATMAP_GRAPH,

    COMPARISON_GRAPH,

    RESULT_CSV,

    RESULT_GRAPH,

    PREDICTION_CSV,

    PREDICTION_GRAPH,

    actual_normal_graph,

    category_graph
]


for number, file_path in enumerate(
    generated_files,
    start=1
):

    print(
        f"{number}. {file_path}"
    )


print("\n" + "=" * 75)
print("PROJECT EXECUTION COMPLETED")
print("=" * 75)
```
