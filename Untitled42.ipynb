```python
import os
import json
import yaml
import pickle
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping


# ======================================================================
# STEP 2: PROJECT PATHS
# ======================================================================

PROJECT_DIR = r"C:\Users\sagni\Downloads\AI-Based District Rainfall Deviation and Anomaly Analysis"

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "Rainfalldistrict_2.csv"
)

H5_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_model.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "rainfall_metadata.json"
)


# Create project directory if it does not exist
os.makedirs(PROJECT_DIR, exist_ok=True)


print("=" * 75)
print("AI-BASED DISTRICT RAINFALL DEVIATION AND ANOMALY ANALYSIS")
print("=" * 75)

print("\nProject Directory:")
print(PROJECT_DIR)

print("\nDataset Path:")
print(CSV_PATH)


# ======================================================================
# STEP 3: CHECK DATASET
# ======================================================================

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}\n\n"
        "Please check the filename and folder location."
    )

print("\nDataset found successfully.")


# ======================================================================
# STEP 4: LOAD DATASET
# ======================================================================

df = pd.read_csv(CSV_PATH)

print("\nDataset loaded successfully.")

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")

print("\nFirst 5 Rows:")
print(df.head())


# ======================================================================
# STEP 5: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace("\n", " ", regex=False)
    .str.replace("\r", " ", regex=False)
    .str.replace(r"\s+", " ", regex=True)
)

print("\nCleaned Column Names:")

for col in df.columns:
    print(col)


# ======================================================================
# STEP 6: REMOVE COMPLETELY EMPTY ROWS/COLUMNS
# ======================================================================

df = df.dropna(axis=0, how="all")
df = df.dropna(axis=1, how="all")

df = df.reset_index(drop=True)

print("\nShape after removing empty rows/columns:")
print(df.shape)


# ======================================================================
# STEP 7: IDENTIFY DISTRICT COLUMN
# ======================================================================

district_column = None

for col in df.columns:

    col_lower = col.lower()

    if "district" in col_lower:
        district_column = col
        break


if district_column is None:

    # Usually first column contains district names
    district_column = df.columns[0]


print("\nDistrict Column:")
print(district_column)


# ======================================================================
# STEP 8: CONVERT RAINFALL COLUMNS TO NUMERIC
# ======================================================================

for col in df.columns:

    if col == district_column:
        continue

    # Remove commas and other unwanted symbols
    cleaned = (
        df[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("-", "", regex=False)
        .str.strip()
    )

    numeric_version = pd.to_numeric(
        cleaned,
        errors="coerce"
    )

    # Replace column if it contains useful numeric values
    if numeric_version.notna().sum() > 0:
        df[col] = numeric_version


# ======================================================================
# STEP 9: FIND NUMERICAL COLUMNS
# ======================================================================

numeric_columns = df.select_dtypes(
    include=[np.number]
).columns.tolist()

print("\nNumeric Columns:")

for col in numeric_columns:
    print(col)


if len(numeric_columns) < 2:
    raise ValueError(
        "Not enough numeric rainfall columns were detected."
    )


# ======================================================================
# STEP 10: HANDLE MISSING VALUES
# ======================================================================

for col in numeric_columns:

    if df[col].isna().any():

        median_value = df[col].median()

        if pd.isna(median_value):
            median_value = 0

        df[col] = df[col].fillna(median_value)


print("\nMissing numeric values handled.")


# ======================================================================
# STEP 11: IDENTIFY ACTUAL AND NORMAL RAINFALL COLUMNS
# ======================================================================

actual_columns = []
normal_columns = []

for col in numeric_columns:

    name = col.lower()

    if "actual" in name:
        actual_columns.append(col)

    if "normal" in name:
        normal_columns.append(col)


print("\nActual Rainfall Columns:")

for col in actual_columns:
    print(col)


print("\nNormal Rainfall Columns:")

for col in normal_columns:
    print(col)


# ======================================================================
# STEP 12: FALLBACK IF COLUMN NAMES DO NOT CONTAIN ACTUAL/NORMAL
# ======================================================================

# Some government rainfall datasets use repeating pairs:
#
# Actual | Normal
# Actual | Normal
# Actual | Normal
#
# If names were lost during CSV conversion, use alternating numeric
# columns as fallback.

if len(actual_columns) == 0 or len(normal_columns) == 0:

    print(
        "\nWARNING:"
        "\nCould not identify Actual/Normal columns from their names."
        "\nUsing alternating numeric columns as fallback."
    )

    actual_columns = numeric_columns[0::2]
    normal_columns = numeric_columns[1::2]


print("\nFinal Actual Columns:")
print(actual_columns)

print("\nFinal Normal Columns:")
print(normal_columns)


# ======================================================================
# STEP 13: MATCH NUMBER OF ACTUAL/NORMAL COLUMNS
# ======================================================================

pair_count = min(
    len(actual_columns),
    len(normal_columns)
)

if pair_count == 0:

    raise ValueError(
        "Actual and Normal rainfall columns could not be identified."
    )


actual_columns = actual_columns[:pair_count]
normal_columns = normal_columns[:pair_count]


# ======================================================================
# STEP 14: CALCULATE TOTAL ACTUAL RAINFALL
# ======================================================================

df["Calculated_Total_Actual"] = df[
    actual_columns
].sum(axis=1)


# ======================================================================
# STEP 15: CALCULATE TOTAL NORMAL RAINFALL
# ======================================================================

df["Calculated_Total_Normal"] = df[
    normal_columns
].sum(axis=1)


# ======================================================================
# STEP 16: CALCULATE RAINFALL DEVIATION
# ======================================================================

df["Rainfall_Deviation_Percentage"] = np.where(

    df["Calculated_Total_Normal"] != 0,

    (
        (
            df["Calculated_Total_Actual"]
            -
            df["Calculated_Total_Normal"]
        )
        /
        df["Calculated_Total_Normal"]
    )
    * 100,

    0
)


print("\nRainfall Deviation Calculated.")

print(
    df[
        [
            district_column,
            "Calculated_Total_Actual",
            "Calculated_Total_Normal",
            "Rainfall_Deviation_Percentage"
        ]
    ].head()
)


# ======================================================================
# STEP 17: CREATE RAINFALL CATEGORY
# ======================================================================

# Simple project classification:
#
# Deficient : deviation < -20%
# Normal    : -20% to +20%
# Excess    : deviation > +20%


def rainfall_category(deviation):

    if deviation < -20:
        return "Deficient"

    elif deviation <= 20:
        return "Normal"

    else:
        return "Excess"


df["Rainfall_Category"] = df[
    "Rainfall_Deviation_Percentage"
].apply(rainfall_category)


print("\nRainfall Category Distribution:")
print(df["Rainfall_Category"].value_counts())


# ======================================================================
# STEP 18: PREPARE MACHINE LEARNING FEATURES
# ======================================================================

feature_columns = actual_columns + normal_columns

# Remove duplicates while preserving order
feature_columns = list(
    dict.fromkeys(feature_columns)
)


X = df[feature_columns].copy()

y_text = df["Rainfall_Category"].copy()


print("\nNumber of Features:")
print(len(feature_columns))

print("\nFeatures:")

for feature in feature_columns:
    print(feature)


# ======================================================================
# STEP 19: ENCODE TARGET LABELS
# ======================================================================

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(y_text)


print("\nTarget Classes:")

for index, class_name in enumerate(label_encoder.classes_):
    print(index, "=", class_name)


# ======================================================================
# STEP 20: STANDARDIZE FEATURES
# ======================================================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


print("\nFeature scaling completed.")


# ======================================================================
# STEP 21: TRAIN / TEST SPLIT
# ======================================================================

# Dataset is relatively small, so use 25% for testing.

try:

    X_train, X_test, y_train, y_test = train_test_split(

        X_scaled,
        y,

        test_size=0.25,
        random_state=42,

        stratify=y
    )

except ValueError:

    # Small class counts can make stratification impossible.

    print(
        "\nStratified split not possible due to small class size."
        "\nUsing normal random split."
    )

    X_train, X_test, y_train, y_test = train_test_split(

        X_scaled,
        y,

        test_size=0.25,
        random_state=42
    )


print("\nTraining Samples:")
print(len(X_train))

print("\nTesting Samples:")
print(len(X_test))


# ======================================================================
# STEP 22: RANDOM FOREST MODEL
# ======================================================================

print("\n" + "=" * 75)
print("TRAINING RANDOM FOREST")
print("=" * 75)


rf_model = RandomForestClassifier(

    n_estimators=200,

    max_depth=None,

    min_samples_split=2,

    min_samples_leaf=1,

    random_state=42,

    class_weight="balanced"
)


rf_model.fit(
    X_train,
    y_train
)


rf_predictions = rf_model.predict(
    X_test
)


rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)


print("\nRandom Forest Accuracy:")
print(round(rf_accuracy * 100, 2), "%")


print("\nRandom Forest Classification Report:")

print(
    classification_report(
        y_test,
        rf_predictions,
        zero_division=0
    )
)


# ======================================================================
# STEP 23: BUILD NEURAL NETWORK
# ======================================================================

print("\n" + "=" * 75)
print("TRAINING NEURAL NETWORK")
print("=" * 75)


number_of_features = X_train.shape[1]

number_of_classes = len(
    label_encoder.classes_
)


nn_model = Sequential([

    Input(
        shape=(number_of_features,)
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.10
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_of_classes,
        activation="softmax"
    )

])


# ======================================================================
# STEP 24: COMPILE NEURAL NETWORK
# ======================================================================

nn_model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


print("\nNeural Network Architecture:")

nn_model.summary()


# ======================================================================
# STEP 25: EARLY STOPPING
# ======================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=20,

    restore_best_weights=True,

    verbose=1
)


# ======================================================================
# STEP 26: TRAIN NEURAL NETWORK
# ======================================================================

history = nn_model.fit(

    X_train,
    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# ======================================================================
# STEP 27: TEST NEURAL NETWORK
# ======================================================================

nn_loss, nn_accuracy = nn_model.evaluate(

    X_test,
    y_test,

    verbose=0
)


print("\nNeural Network Test Accuracy:")
print(round(nn_accuracy * 100, 2), "%")


# ======================================================================
# STEP 28: NEURAL NETWORK PREDICTIONS
# ======================================================================

nn_probabilities = nn_model.predict(
    X_test,
    verbose=0
)


nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)


print("\nNeural Network Classification Report:")

print(
    classification_report(
        y_test,
        nn_predictions,
        zero_division=0
    )
)


# ======================================================================
# STEP 29: SAVE H5 MODEL
# ======================================================================

print("\n" + "=" * 75)
print("SAVING FILES")
print("=" * 75)


nn_model.save(
    H5_PATH
)


print("\nH5 Model Saved:")
print(H5_PATH)


# ======================================================================
# STEP 30: SAVE PKL FILE
# ======================================================================

# Store everything needed for classical ML prediction.

pkl_data = {

    "random_forest_model": rf_model,

    "scaler": scaler,

    "label_encoder": label_encoder,

    "feature_columns": feature_columns,

    "actual_columns": actual_columns,

    "normal_columns": normal_columns,

    "district_column": district_column,

    "classification_thresholds": {
        "deficient_below": -20,
        "normal_lower": -20,
        "normal_upper": 20,
        "excess_above": 20
    }
}


with open(
    PKL_PATH,
    "wb"
) as file:

    pickle.dump(
        pkl_data,
        file
    )


print("\nPKL File Saved:")
print(PKL_PATH)


# ======================================================================
# STEP 31: SAVE YAML CONFIGURATION
# ======================================================================

yaml_data = {

    "project": {
        "name":
            "AI-Based District Rainfall Deviation and Anomaly Analysis",

        "dataset":
            "Rainfalldistrict_2.csv",

        "problem_type":
            "Multiclass Classification",

        "target":
            "Rainfall_Category"
    },

    "dataset": {
        "rows":
            int(df.shape[0]),

        "columns":
            int(df.shape[1]),

        "district_column":
            str(district_column)
    },

    "features": {
        "actual_columns":
            [str(x) for x in actual_columns],

        "normal_columns":
            [str(x) for x in normal_columns],

        "model_features":
            [str(x) for x in feature_columns]
    },

    "rainfall_categories": {

        "Deficient":
            "Deviation below -20%",

        "Normal":
            "Deviation from -20% to +20%",

        "Excess":
            "Deviation above +20%"
    },

    "models": {

        "random_forest": {
            "estimators": 200,
            "random_state": 42,
            "accuracy":
                float(rf_accuracy)
        },

        "neural_network": {
            "epochs": 200,
            "batch_size": 8,
            "optimizer": "adam",
            "accuracy":
                float(nn_accuracy)
        }
    },

    "output_files": {
        "h5":
            "rainfall_model.h5",

        "pkl":
            "rainfall_model.pkl",

        "yaml":
            "rainfall_config.yaml",

        "json":
            "rainfall_metadata.json"
    }
}


with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        yaml_data,
        file,
        sort_keys=False,
        allow_unicode=True
    )


print("\nYAML File Saved:")
print(YAML_PATH)


# ======================================================================
# STEP 32: SAVE JSON METADATA
# ======================================================================

json_data = {

    "project_name":
        "AI-Based District Rainfall Deviation and Anomaly Analysis",

    "dataset":
        "Rainfalldistrict_2.csv",

    "dataset_rows":
        int(df.shape[0]),

    "dataset_columns":
        int(df.shape[1]),

    "district_column":
        str(district_column),

    "actual_rainfall_columns":
        [str(x) for x in actual_columns],

    "normal_rainfall_columns":
        [str(x) for x in normal_columns],

    "feature_columns":
        [str(x) for x in feature_columns],

    "target":
        "Rainfall_Category",

    "classes":
        [
            str(x)
            for x in label_encoder.classes_
        ],

    "rainfall_thresholds": {

        "deficient":
            "< -20%",

        "normal":
            "-20% to +20%",

        "excess":
            "> +20%"
    },

    "random_forest_accuracy":
        float(rf_accuracy),

    "neural_network_accuracy":
        float(nn_accuracy),

    "random_forest_confusion_matrix":
        confusion_matrix(
            y_test,
            rf_predictions
        ).tolist(),

    "neural_network_confusion_matrix":
        confusion_matrix(
            y_test,
            nn_predictions
        ).tolist(),

    "saved_files": {

        "h5":
            H5_PATH,

        "pkl":
            PKL_PATH,

        "yaml":
            YAML_PATH,

        "json":
            JSON_PATH
    }
}


with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        json_data,
        file,
        indent=4,
        ensure_ascii=False
    )


print("\nJSON File Saved:")
print(JSON_PATH)


# ======================================================================
# STEP 33: TEST PKL FILE
# ======================================================================

print("\n" + "=" * 75)
print("TESTING SAVED PKL FILE")
print("=" * 75)


with open(
    PKL_PATH,
    "rb"
) as file:

    loaded_pkl = pickle.load(file)


loaded_rf_model = loaded_pkl[
    "random_forest_model"
]

loaded_scaler = loaded_pkl[
    "scaler"
]


sample = df[
    feature_columns
].iloc[[0]]


sample_scaled = loaded_scaler.transform(
    sample
)


sample_prediction = loaded_rf_model.predict(
    sample_scaled
)


sample_category = label_encoder.inverse_transform(
    sample_prediction
)[0]


print("\nSample District:")
print(df.iloc[0][district_column])

print("\nPKL Model Prediction:")
print(sample_category)


# ======================================================================
# STEP 34: TEST H5 MODEL
# ======================================================================

print("\n" + "=" * 75)
print("TESTING SAVED H5 MODEL")
print("=" * 75)


loaded_nn_model = tf.keras.models.load_model(
    H5_PATH
)


sample_nn_probability = loaded_nn_model.predict(
    sample_scaled,
    verbose=0
)


sample_nn_prediction = np.argmax(
    sample_nn_probability,
    axis=1
)


sample_nn_category = label_encoder.inverse_transform(
    sample_nn_prediction
)[0]


print("\nH5 Model Prediction:")
print(sample_nn_category)


# ======================================================================
# STEP 35: FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 75)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 75)


print("\nRandom Forest Accuracy:")
print(
    f"{rf_accuracy * 100:.2f}%"
)


print("\nNeural Network Accuracy:")
print(
    f"{nn_accuracy * 100:.2f}%"
)


print("\nGenerated Files:")

print(
    "1.",
    H5_PATH
)

print(
    "2.",
    PKL_PATH
)

print(
    "3.",
    YAML_PATH
)

print(
    "4.",
    JSON_PATH
)


print("\nAll requested files have been generated successfully.")
```
