```python
import os
import json
import yaml
import pickle
import random
import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    cross_val_score,
    StratifiedKFold
)

from sklearn.preprocessing import (
    StandardScaler,
    LabelEncoder
)

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)

import tensorflow as tf

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Dense,
    Dropout,
    Input
)

from tensorflow.keras.callbacks import (
    EarlyStopping
)


# ======================================================================
# STEP 2: REPRODUCIBILITY
# ======================================================================

SEED = 42

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)


# ======================================================================
# STEP 3: PROJECT DIRECTORY
# ======================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based District Rainfall Deviation and Anomaly Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "Rainfalldistrict_2.csv"
)

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# ======================================================================
# STEP 4: OUTPUT PATHS
# ======================================================================

AIS_H5_PATH = os.path.join(
    PROJECT_DIR,
    "ais_rainfall_model.h5"
)

AIS_PKL_PATH = os.path.join(
    PROJECT_DIR,
    "ais_rainfall_model.pkl"
)

AIS_YAML_PATH = os.path.join(
    PROJECT_DIR,
    "ais_config.yaml"
)

AIS_JSON_PATH = os.path.join(
    PROJECT_DIR,
    "ais_metadata.json"
)


AIS_ACCURACY_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_accuracy_graph.png"
)

AIS_HEATMAP_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_heatmap.png"
)

AIS_COMPARISON_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_comparison_graph.png"
)


AIS_RESULT_CSV = os.path.join(
    PROJECT_DIR,
    "ais_result.csv"
)

AIS_RESULT_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_result_graph.png"
)


AIS_PREDICTION_CSV = os.path.join(
    PROJECT_DIR,
    "ais_prediction.csv"
)

AIS_PREDICTION_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_prediction_graph.png"
)


AIS_FEATURE_SELECTION_CSV = os.path.join(
    PROJECT_DIR,
    "ais_feature_selection.csv"
)

AIS_FEATURE_SELECTION_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_feature_selection_graph.png"
)


AIS_ACTUAL_NORMAL_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_actual_vs_normal_rainfall.png"
)

AIS_CATEGORY_GRAPH = os.path.join(
    PROJECT_DIR,
    "ais_category_distribution.png"
)


# ======================================================================
# STEP 5: DISPLAY PROJECT INFORMATION
# ======================================================================

print("=" * 80)

print(
    "AI-BASED DISTRICT RAINFALL DEVIATION AND ANOMALY ANALYSIS"
)

print(
    "ARTIFICIAL IMMUNE SYSTEM (AIS) OPTIMIZATION"
)

print("=" * 80)

print("\nProject Directory:")
print(PROJECT_DIR)

print("\nDataset:")
print(CSV_PATH)


# ======================================================================
# STEP 6: CHECK DATASET
# ======================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}"
    )


print("\nDataset found successfully.")


# ======================================================================
# STEP 7: LOAD DATASET
# ======================================================================

df = pd.read_csv(
    CSV_PATH
)

print("\nDataset loaded successfully.")

print(
    "\nOriginal Dataset Shape:",
    df.shape
)


# ======================================================================
# STEP 8: CLEAN COLUMN NAMES
# ======================================================================

df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.replace(
        "\n",
        " ",
        regex=False
    )
    .str.replace(
        "\r",
        " ",
        regex=False
    )
    .str.replace(
        r"\s+",
        " ",
        regex=True
    )
)


# ======================================================================
# STEP 9: REMOVE EMPTY ROWS/COLUMNS
# ======================================================================

df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)

df.reset_index(
    drop=True,
    inplace=True
)


print(
    "\nDataset Shape After Cleaning:",
    df.shape
)


print("\nColumns:")

for i, col in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i}. {col}"
    )


# ======================================================================
# STEP 10: IDENTIFY DISTRICT COLUMN
# ======================================================================

district_column = None

for col in df.columns:

    if "district" in col.lower():

        district_column = col

        break


if district_column is None:

    district_column = df.columns[0]


print("\nDistrict Column:")
print(district_column)


# ======================================================================
# STEP 11: CLEAN DISTRICT NAMES
# ======================================================================

df[district_column] = (
    df[district_column]
    .astype(str)
    .str.strip()
)


invalid_names = [
    "",
    "nan",
    "none",
    "null"
]


df = df[
    ~df[district_column]
    .str.lower()
    .isin(invalid_names)
].copy()


df.reset_index(
    drop=True,
    inplace=True
)


# ======================================================================
# STEP 12: CONVERT RAINFALL DATA TO NUMERIC
# ======================================================================

for col in df.columns:

    if col == district_column:
        continue

    cleaned = (
        df[col]
        .astype(str)
        .str.replace(
            ",",
            "",
            regex=False
        )
        .str.strip()
    )


    cleaned = cleaned.replace(

        [
            "",
            "-",
            "--",
            "NA",
            "N/A",
            "nan",
            "None",
            "null"
        ],

        np.nan
    )


    numeric_version = pd.to_numeric(

        cleaned,

        errors="coerce"
    )


    if numeric_version.notna().sum() > 0:

        df[col] = numeric_version


# ======================================================================
# STEP 13: FIND NUMERICAL COLUMNS
# ======================================================================

numeric_columns = df.select_dtypes(

    include=[
        np.number
    ]

).columns.tolist()


print("\nNumeric Columns:")

for col in numeric_columns:

    print(col)


if len(numeric_columns) < 2:

    raise ValueError(
        "Not enough numerical rainfall columns were detected."
    )


# ======================================================================
# STEP 14: HANDLE MISSING VALUES
# ======================================================================

for col in numeric_columns:

    median_value = df[col].median()

    if pd.isna(median_value):

        median_value = 0


    df[col] = df[col].fillna(
        median_value
    )


print("\nMissing values handled successfully.")


# ======================================================================
# STEP 15: IDENTIFY ACTUAL AND NORMAL COLUMNS
# ======================================================================

actual_columns = []

normal_columns = []


for col in numeric_columns:

    lower = col.lower()


    if "actual" in lower:

        actual_columns.append(
            col
        )


    elif "normal" in lower:

        normal_columns.append(
            col
        )


# ======================================================================
# STEP 16: FALLBACK COLUMN PAIRING
# ======================================================================

if (
    len(actual_columns) == 0
    or
    len(normal_columns) == 0
):

    print(
        "\nActual/Normal names were not detected."
    )

    print(
        "Using alternating numerical columns."
    )


    actual_columns = (
        numeric_columns[0::2]
    )

    normal_columns = (
        numeric_columns[1::2]
    )


# ======================================================================
# STEP 17: MATCH COLUMN PAIRS
# ======================================================================

pair_count = min(

    len(actual_columns),

    len(normal_columns)
)


if pair_count == 0:

    raise ValueError(
        "Could not identify Actual and Normal rainfall columns."
    )


actual_columns = (
    actual_columns[:pair_count]
)

normal_columns = (
    normal_columns[:pair_count]
)


print("\nActual Columns:")

for col in actual_columns:

    print(col)


print("\nNormal Columns:")

for col in normal_columns:

    print(col)


# ======================================================================
# STEP 18: TOTAL ACTUAL RAINFALL
# ======================================================================

df[
    "Calculated_Total_Actual"
] = df[
    actual_columns
].sum(
    axis=1
)


# ======================================================================
# STEP 19: TOTAL NORMAL RAINFALL
# ======================================================================

df[
    "Calculated_Total_Normal"
] = df[
    normal_columns
].sum(
    axis=1
)


# ======================================================================
# STEP 20: CALCULATE DEVIATION
# ======================================================================

df[
    "Rainfall_Deviation_Percentage"
] = np.where(

    df[
        "Calculated_Total_Normal"
    ] != 0,

    (

        (

            df[
                "Calculated_Total_Actual"
            ]

            -

            df[
                "Calculated_Total_Normal"
            ]

        )

        /

        df[
            "Calculated_Total_Normal"
        ]

    )

    * 100,

    0
)


# ======================================================================
# STEP 21: CREATE TARGET CATEGORY
# ======================================================================

def rainfall_category(
    deviation
):

    if deviation < -20:

        return "Deficient"

    elif deviation <= 20:

        return "Normal"

    else:

        return "Excess"


df[
    "Rainfall_Category"
] = df[
    "Rainfall_Deviation_Percentage"
].apply(
    rainfall_category
)


print("\nRainfall Category Distribution:")

print(
    df[
        "Rainfall_Category"
    ].value_counts()
)


# ======================================================================
# STEP 22: PREPARE ORIGINAL FEATURES
# ======================================================================

feature_columns = list(

    dict.fromkeys(

        actual_columns
        +
        normal_columns
    )
)


X_original = df[
    feature_columns
].copy()


y_text = df[
    "Rainfall_Category"
].copy()


print("\nOriginal Number of Features:")

print(
    len(feature_columns)
)


print("\nOriginal Features:")

for feature in feature_columns:

    print(feature)


# ======================================================================
# STEP 23: ENCODE TARGET
# ======================================================================

label_encoder = LabelEncoder()


y = label_encoder.fit_transform(
    y_text
)


print("\nEncoded Classes:")

for index, class_name in enumerate(
    label_encoder.classes_
):

    print(
        index,
        "=",
        class_name
    )


# ======================================================================
# STEP 24: SPLIT DATA BEFORE AIS
# ======================================================================
# AIS evaluates features using training data only.
# This prevents direct use of the final test set during optimization.
# ======================================================================

indices = np.arange(
    len(df)
)


try:

    (
        train_indices,
        test_indices
    ) = train_test_split(

        indices,

        test_size=0.25,

        random_state=SEED,

        stratify=y
    )


except ValueError:

    print(
        "\nStratified split unavailable."
    )

    print(
        "Using standard random split."
    )


    (
        train_indices,
        test_indices
    ) = train_test_split(

        indices,

        test_size=0.25,

        random_state=SEED
    )


X_train_original = (
    X_original.iloc[
        train_indices
    ].reset_index(
        drop=True
    )
)


X_test_original = (
    X_original.iloc[
        test_indices
    ].reset_index(
        drop=True
    )
)


y_train = y[
    train_indices
]


y_test = y[
    test_indices
]


print("\nTraining Samples:")

print(
    len(X_train_original)
)


print("\nTesting Samples:")

print(
    len(X_test_original)
)


# ======================================================================
# STEP 25: AIS PARAMETERS
# ======================================================================

AIS_POPULATION_SIZE = 30

AIS_GENERATIONS = 25

AIS_CLONE_FACTOR = 5

AIS_MUTATION_RATE = 0.15

AIS_ELITE_SIZE = 6


number_of_features = len(
    feature_columns
)


print("\n" + "=" * 80)

print(
    "STARTING ARTIFICIAL IMMUNE SYSTEM FEATURE SELECTION"
)

print("=" * 80)


print(
    "\nPopulation Size:",
    AIS_POPULATION_SIZE
)

print(
    "Generations:",
    AIS_GENERATIONS
)

print(
    "Clone Factor:",
    AIS_CLONE_FACTOR
)

print(
    "Mutation Rate:",
    AIS_MUTATION_RATE
)

print(
    "Elite Size:",
    AIS_ELITE_SIZE
)


# ======================================================================
# STEP 26: CREATE RANDOM ANTIBODY
# ======================================================================
# 1 = feature selected
# 0 = feature not selected
# ======================================================================

def create_antibody():

    antibody = np.random.randint(

        0,
        2,

        size=number_of_features
    )


    # At least one feature must be selected

    if np.sum(antibody) == 0:

        random_position = np.random.randint(
            0,
            number_of_features
        )

        antibody[
            random_position
        ] = 1


    return antibody


# ======================================================================
# STEP 27: AIS FITNESS FUNCTION
# ======================================================================
# Fitness:
# Random Forest cross-validation accuracy
# minus a small penalty for selecting too many features.
#
# This encourages good accuracy with fewer features.
# ======================================================================

def calculate_fitness(
    antibody
):

    selected_indices = np.where(
        antibody == 1
    )[0]


    if len(selected_indices) == 0:

        return 0.0


    selected_features = [

        feature_columns[i]

        for i in selected_indices
    ]


    X_selected = X_train_original[
        selected_features
    ]


    model = RandomForestClassifier(

        n_estimators=100,

        random_state=SEED,

        class_weight="balanced"
    )


    # --------------------------------------------------------------
    # Determine safe CV folds for small dataset
    # --------------------------------------------------------------

    unique_classes, class_counts = np.unique(

        y_train,

        return_counts=True
    )


    minimum_class_count = (
        class_counts.min()
    )


    cv_folds = min(
        3,
        minimum_class_count
    )


    # --------------------------------------------------------------
    # If cross-validation is possible
    # --------------------------------------------------------------

    if cv_folds >= 2:

        cv = StratifiedKFold(

            n_splits=cv_folds,

            shuffle=True,

            random_state=SEED
        )


        try:

            scores = cross_val_score(

                model,

                X_selected,

                y_train,

                cv=cv,

                scoring="accuracy"
            )


            accuracy = (
                scores.mean()
            )


        except Exception:

            model.fit(
                X_selected,
                y_train
            )

            predictions = model.predict(
                X_selected
            )

            accuracy = accuracy_score(
                y_train,
                predictions
            )


    else:

        model.fit(
            X_selected,
            y_train
        )

        predictions = model.predict(
            X_selected
        )

        accuracy = accuracy_score(
            y_train,
            predictions
        )


    # --------------------------------------------------------------
    # Feature penalty
    # --------------------------------------------------------------

    feature_ratio = (

        len(selected_features)

        /

        number_of_features
    )


    penalty = (
        0.02
        *
        feature_ratio
    )


    fitness = (
        accuracy
        -
        penalty
    )


    return float(
        fitness
    )


# ======================================================================
# STEP 28: MUTATION FUNCTION
# ======================================================================

def mutate_antibody(
    antibody,
    mutation_rate
):

    mutated = antibody.copy()


    for i in range(
        len(mutated)
    ):

        if np.random.rand() < mutation_rate:

            mutated[i] = (
                1
                -
                mutated[i]
            )


    if np.sum(mutated) == 0:

        random_position = np.random.randint(

            0,

            len(mutated)
        )

        mutated[
            random_position
        ] = 1


    return mutated


# ======================================================================
# STEP 29: INITIAL AIS POPULATION
# ======================================================================

population = [

    create_antibody()

    for _ in range(
        AIS_POPULATION_SIZE
    )
]


best_antibody = None

best_fitness = -np.inf


ais_history = []


# ======================================================================
# STEP 30: AIS OPTIMIZATION LOOP
# ======================================================================

for generation in range(
    AIS_GENERATIONS
):

    print(
        "\nGeneration",
        generation + 1,
        "/",
        AIS_GENERATIONS
    )


    # --------------------------------------------------------------
    # Calculate fitness
    # --------------------------------------------------------------

    fitness_values = np.array([

        calculate_fitness(
            antibody
        )

        for antibody in population
    ])


    # --------------------------------------------------------------
    # Sort population
    # --------------------------------------------------------------

    sorted_indices = np.argsort(
        fitness_values
    )[::-1]


    population = [

        population[i]

        for i in sorted_indices
    ]


    fitness_values = (
        fitness_values[
            sorted_indices
        ]
    )


    # --------------------------------------------------------------
    # Best generation antibody
    # --------------------------------------------------------------

    generation_best = (
        population[0].copy()
    )


    generation_best_fitness = (
        fitness_values[0]
    )


    # --------------------------------------------------------------
    # Global best
    # --------------------------------------------------------------

    if (
        generation_best_fitness
        >
        best_fitness
    ):

        best_fitness = (
            generation_best_fitness
        )

        best_antibody = (
            generation_best.copy()
        )


    ais_history.append(
        best_fitness
    )


    selected_count = int(
        np.sum(
            generation_best
        )
    )


    print(
        "Best Fitness:",
        round(
            best_fitness,
            4
        )
    )


    print(
        "Selected Features:",
        selected_count,
        "/",
        number_of_features
    )


    # --------------------------------------------------------------
    # Elite selection
    # --------------------------------------------------------------

    elite_population = [

        population[i].copy()

        for i in range(
            min(
                AIS_ELITE_SIZE,
                len(population)
            )
        )
    ]


    elite_fitness = fitness_values[
        :len(
            elite_population
        )
    ]


    # --------------------------------------------------------------
    # Clonal selection
    # --------------------------------------------------------------

    clones = []


    for elite_index, elite in enumerate(
        elite_population
    ):

        relative_rank = (

            elite_index + 1

        )


        number_of_clones = max(

            1,

            int(

                AIS_CLONE_FACTOR

                *

                (
                    AIS_ELITE_SIZE
                    /
                    relative_rank
                )

            )
        )


        # Better antibodies get lower mutation

        adaptive_mutation_rate = (

            AIS_MUTATION_RATE

            *

            (
                relative_rank
                /
                AIS_ELITE_SIZE
            )
        )


        for _ in range(
            number_of_clones
        ):

            clone = mutate_antibody(

                elite,

                adaptive_mutation_rate
            )


            clones.append(
                clone
            )


    # --------------------------------------------------------------
    # Evaluate clones
    # --------------------------------------------------------------

    clone_fitness = np.array([

        calculate_fitness(
            clone
        )

        for clone in clones
    ])


    # --------------------------------------------------------------
    # Combine elite + clones
    # --------------------------------------------------------------

    combined_population = (
        elite_population
        +
        clones
    )


    combined_fitness = np.concatenate(

        [
            elite_fitness,
            clone_fitness
        ]
    )


    combined_indices = np.argsort(

        combined_fitness

    )[::-1]


    # --------------------------------------------------------------
    # Keep best antibodies
    # --------------------------------------------------------------

    next_population = []


    keep_count = min(

        AIS_POPULATION_SIZE
        -
        5,

        len(
            combined_indices
        )
    )


    for idx in combined_indices[
        :keep_count
    ]:

        next_population.append(

            combined_population[
                idx
            ].copy()
        )


    # --------------------------------------------------------------
    # Add diversity
    # --------------------------------------------------------------

    while len(
        next_population
    ) < AIS_POPULATION_SIZE:

        next_population.append(

            create_antibody()
        )


    population = (
        next_population
    )


# ======================================================================
# STEP 31: AIS FINAL RESULT
# ======================================================================

print("\n" + "=" * 80)

print(
    "AIS FEATURE SELECTION COMPLETED"
)

print("=" * 80)


selected_indices = np.where(
    best_antibody == 1
)[0]


selected_features = [

    feature_columns[i]

    for i in selected_indices
]


print("\nBest AIS Fitness:")

print(
    round(
        best_fitness,
        4
    )
)


print(
    "\nSelected Features:",
    len(
        selected_features
    )
)


for feature in selected_features:

    print(
        "✓",
        feature
    )


# ======================================================================
# STEP 32: CREATE FEATURE SELECTION CSV
# ======================================================================

feature_selection_df = pd.DataFrame({

    "Feature":
        feature_columns,

    "Selected_By_AIS":
        [
            "Yes" if value == 1 else "No"

            for value in best_antibody
        ],

    "Binary_Value":
        best_antibody.astype(
            int
        )
})


feature_selection_df.to_csv(

    AIS_FEATURE_SELECTION_CSV,

    index=False
)


print(
    "\nSaved:",
    AIS_FEATURE_SELECTION_CSV
)


# ======================================================================
# STEP 33: AIS OPTIMIZATION GRAPH
# ======================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    range(
        1,
        AIS_GENERATIONS + 1
    ),

    ais_history,

    marker="o"
)


plt.xlabel(
    "AIS Generation"
)

plt.ylabel(
    "Best Fitness"
)

plt.title(
    "AIS Feature Selection Optimization"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_FEATURE_SELECTION_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_FEATURE_SELECTION_GRAPH
)


# ======================================================================
# STEP 34: CREATE AIS SELECTED DATA
# ======================================================================

X_selected = df[
    selected_features
].copy()


X_train_selected = X_selected.iloc[
    train_indices
].reset_index(
    drop=True
)


X_test_selected = X_selected.iloc[
    test_indices
].reset_index(
    drop=True
)


# ======================================================================
# STEP 35: SCALE AIS FEATURES
# ======================================================================

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(
    X_train_selected
)


X_test_scaled = scaler.transform(
    X_test_selected
)


X_full_scaled = scaler.transform(
    X_selected
)


# ======================================================================
# STEP 36: TRAIN AIS RANDOM FOREST
# ======================================================================

print("\n" + "=" * 80)

print(
    "TRAINING RANDOM FOREST WITH AIS FEATURES"
)

print("=" * 80)


rf_model = RandomForestClassifier(

    n_estimators=300,

    random_state=SEED,

    class_weight="balanced"
)


rf_model.fit(

    X_train_scaled,

    y_train
)


rf_prediction = rf_model.predict(
    X_test_scaled
)


rf_probability = rf_model.predict_proba(
    X_test_scaled
)


# ======================================================================
# STEP 37: RANDOM FOREST METRICS
# ======================================================================

rf_accuracy = accuracy_score(

    y_test,

    rf_prediction
)


rf_precision = precision_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0
)


rf_recall = recall_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0
)


rf_f1 = f1_score(

    y_test,

    rf_prediction,

    average="weighted",

    zero_division=0
)


print(
    f"\nAIS Random Forest Accuracy: "
    f"{rf_accuracy * 100:.2f}%"
)


print("\nClassification Report:")


print(

    classification_report(

        y_test,

        rf_prediction,

        zero_division=0
    )

)


# ======================================================================
# STEP 38: BUILD AIS NEURAL NETWORK
# ======================================================================

print("\n" + "=" * 80)

print(
    "TRAINING NEURAL NETWORK WITH AIS FEATURES"
)

print("=" * 80)


number_selected_features = (
    X_train_scaled.shape[1]
)


number_classes = len(
    label_encoder.classes_
)


nn_model = Sequential([

    Input(
        shape=(
            number_selected_features,
        )
    ),

    Dense(
        64,
        activation="relu"
    ),

    Dropout(
        0.20
    ),

    Dense(
        32,
        activation="relu"
    ),

    Dropout(
        0.10
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        number_classes,
        activation="softmax"
    )

])


# ======================================================================
# STEP 39: COMPILE NN
# ======================================================================

nn_model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


# ======================================================================
# STEP 40: EARLY STOPPING
# ======================================================================

early_stop = EarlyStopping(

    monitor="val_loss",

    patience=25,

    restore_best_weights=True
)


# ======================================================================
# STEP 41: TRAIN NEURAL NETWORK
# ======================================================================

history = nn_model.fit(

    X_train_scaled,

    y_train,

    validation_split=0.20,

    epochs=200,

    batch_size=8,

    callbacks=[
        early_stop
    ],

    verbose=1
)


# ======================================================================
# STEP 42: NN TEST PREDICTIONS
# ======================================================================

nn_probability = nn_model.predict(

    X_test_scaled,

    verbose=0
)


nn_prediction = np.argmax(

    nn_probability,

    axis=1
)


# ======================================================================
# STEP 43: NN METRICS
# ======================================================================

nn_accuracy = accuracy_score(

    y_test,

    nn_prediction
)


nn_precision = precision_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0
)


nn_recall = recall_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0
)


nn_f1 = f1_score(

    y_test,

    nn_prediction,

    average="weighted",

    zero_division=0
)


print(
    f"\nAIS Neural Network Accuracy: "
    f"{nn_accuracy * 100:.2f}%"
)


print("\nClassification Report:")


print(

    classification_report(

        y_test,

        nn_prediction,

        zero_division=0
    )

)


# ======================================================================
# STEP 44: SAVE AIS H5 MODEL
# ======================================================================

nn_model.save(
    AIS_H5_PATH
)


print(
    "\nSaved:",
    AIS_H5_PATH
)


# ======================================================================
# STEP 45: SAVE AIS PKL
# ======================================================================

pkl_data = {

    "random_forest_model":
        rf_model,

    "scaler":
        scaler,

    "label_encoder":
        label_encoder,

    "all_features":
        feature_columns,

    "ais_selected_features":
        selected_features,

    "ais_best_antibody":
        best_antibody.tolist(),

    "ais_best_fitness":
        float(
            best_fitness
        ),

    "actual_columns":
        actual_columns,

    "normal_columns":
        normal_columns,

    "district_column":
        district_column,

    "classification_thresholds": {

        "deficient_below":
            -20,

        "normal_lower":
            -20,

        "normal_upper":
            20,

        "excess_above":
            20
    }
}


with open(

    AIS_PKL_PATH,

    "wb"

) as file:

    pickle.dump(

        pkl_data,

        file
    )


print(
    "Saved:",
    AIS_PKL_PATH
)


# ======================================================================
# STEP 46: SAVE AIS YAML
# ======================================================================

yaml_data = {

    "project": {

        "name":
            "AI-Based District Rainfall Deviation and Anomaly Analysis",

        "optimization":
            "Artificial Immune System",

        "dataset":
            "Rainfalldistrict_2.csv"
    },

    "ais": {

        "population_size":
            AIS_POPULATION_SIZE,

        "generations":
            AIS_GENERATIONS,

        "clone_factor":
            AIS_CLONE_FACTOR,

        "mutation_rate":
            AIS_MUTATION_RATE,

        "elite_size":
            AIS_ELITE_SIZE,

        "best_fitness":
            float(
                best_fitness
            ),

        "original_feature_count":
            int(
                len(
                    feature_columns
                )
            ),

        "selected_feature_count":
            int(
                len(
                    selected_features
                )
            ),

        "selected_features":
            [
                str(x)
                for x in selected_features
            ]
    },

    "random_forest": {

        "accuracy":
            float(
                rf_accuracy
            ),

        "precision":
            float(
                rf_precision
            ),

        "recall":
            float(
                rf_recall
            ),

        "f1":
            float(
                rf_f1
            )
    },

    "neural_network": {

        "accuracy":
            float(
                nn_accuracy
            ),

        "precision":
            float(
                nn_precision
            ),

        "recall":
            float(
                nn_recall
            ),

        "f1":
            float(
                nn_f1
            )
    }
}


with open(

    AIS_YAML_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_data,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "Saved:",
    AIS_YAML_PATH
)


# ======================================================================
# STEP 47: SAVE AIS JSON
# ======================================================================

json_data = {

    "project":
        "AI-Based District Rainfall Deviation and Anomaly Analysis",

    "algorithm":
        "Artificial Immune System",

    "dataset":
        "Rainfalldistrict_2.csv",

    "original_features":
        [
            str(x)
            for x in feature_columns
        ],

    "selected_features":
        [
            str(x)
            for x in selected_features
        ],

    "ais_best_antibody":
        [
            int(x)
            for x in best_antibody
        ],

    "ais_best_fitness":
        float(
            best_fitness
        ),

    "random_forest_accuracy":
        float(
            rf_accuracy
        ),

    "random_forest_precision":
        float(
            rf_precision
        ),

    "random_forest_recall":
        float(
            rf_recall
        ),

    "random_forest_f1":
        float(
            rf_f1
        ),

    "neural_network_accuracy":
        float(
            nn_accuracy
        ),

    "neural_network_precision":
        float(
            nn_precision
        ),

    "neural_network_recall":
        float(
            nn_recall
        ),

    "neural_network_f1":
        float(
            nn_f1
        )
}


with open(

    AIS_JSON_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        json_data,

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "Saved:",
    AIS_JSON_PATH
)


# ======================================================================
# STEP 48: AIS ACCURACY GRAPH
# ======================================================================

print(
    "\nCreating AIS accuracy graph..."
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(

    history.history[
        "accuracy"
    ],

    marker="o",

    markersize=3,

    label="Training Accuracy"
)


plt.plot(

    history.history[
        "val_accuracy"
    ],

    marker="s",

    markersize=3,

    label="Validation Accuracy"
)


plt.title(
    "AIS Neural Network Training and Validation Accuracy"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.ylim(
    0,
    1.05
)


plt.grid(
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    AIS_ACCURACY_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_ACCURACY_GRAPH
)


# ======================================================================
# STEP 49: AIS CONFUSION MATRIX HEATMAP
# ======================================================================

print(
    "\nCreating AIS confusion matrix heatmap..."
)


cm = confusion_matrix(

    y_test,

    rf_prediction,

    labels=np.arange(
        len(
            label_encoder.classes_
        )
    )
)


plt.figure(
    figsize=(8, 6)
)


plt.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"
)


plt.title(
    "AIS Random Forest Confusion Matrix"
)


plt.colorbar()


ticks = np.arange(

    len(
        label_encoder.classes_
    )
)


plt.xticks(

    ticks,

    label_encoder.classes_,

    rotation=45
)


plt.yticks(

    ticks,

    label_encoder.classes_
)


threshold = (

    cm.max() / 2

    if cm.size > 0

    else 0
)


for i in range(
    cm.shape[0]
):

    for j in range(
        cm.shape[1]
    ):

        plt.text(

            j,

            i,

            str(
                cm[i, j]
            ),

            ha="center",

            va="center",

            color=(

                "white"

                if cm[i, j] > threshold

                else "black"
            )
        )


plt.xlabel(
    "Predicted Category"
)


plt.ylabel(
    "Actual Category"
)


plt.tight_layout()


plt.savefig(

    AIS_HEATMAP_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_HEATMAP_GRAPH
)


# ======================================================================
# STEP 50: AIS MODEL COMPARISON
# ======================================================================

comparison_df = pd.DataFrame({

    "Metric": [

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score"
    ],

    "AIS Random Forest": [

        rf_accuracy,

        rf_precision,

        rf_recall,

        rf_f1
    ],

    "AIS Neural Network": [

        nn_accuracy,

        nn_precision,

        nn_recall,

        nn_f1
    ]
})


print("\nAIS Model Comparison:")

print(
    comparison_df
)


x = np.arange(
    len(
        comparison_df
    )
)


width = 0.35


plt.figure(
    figsize=(10, 6)
)


bars1 = plt.bar(

    x - width / 2,

    comparison_df[
        "AIS Random Forest"
    ],

    width,

    label="AIS Random Forest"
)


bars2 = plt.bar(

    x + width / 2,

    comparison_df[
        "AIS Neural Network"
    ],

    width,

    label="AIS Neural Network"
)


plt.xticks(

    x,

    comparison_df[
        "Metric"
    ]
)


plt.ylabel(
    "Score"
)


plt.xlabel(
    "Evaluation Metric"
)


plt.title(
    "AIS Optimized Model Comparison"
)


plt.ylim(
    0,
    1.10
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


for bars in [
    bars1,
    bars2
]:

    for bar in bars:

        height = (
            bar.get_height()
        )


        plt.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            height + 0.02,

            f"{height:.2f}",

            ha="center",

            va="bottom"
        )


plt.tight_layout()


plt.savefig(

    AIS_COMPARISON_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_COMPARISON_GRAPH
)


# ======================================================================
# STEP 51: CREATE AIS RESULT CSV
# ======================================================================

ais_result_df = df[

    [
        district_column,

        "Calculated_Total_Actual",

        "Calculated_Total_Normal",

        "Rainfall_Deviation_Percentage",

        "Rainfall_Category"
    ]

].copy()


ais_result_df.columns = [

    "District",

    "Actual_Rainfall",

    "Normal_Rainfall",

    "Deviation_Percentage",

    "Rainfall_Category"
]


ais_result_df[
    "Actual_Rainfall"
] = ais_result_df[
    "Actual_Rainfall"
].round(
    2
)


ais_result_df[
    "Normal_Rainfall"
] = ais_result_df[
    "Normal_Rainfall"
].round(
    2
)


ais_result_df[
    "Deviation_Percentage"
] = ais_result_df[
    "Deviation_Percentage"
].round(
    2
)


ais_result_df.to_csv(

    AIS_RESULT_CSV,

    index=False
)


print(
    "\nSaved:",
    AIS_RESULT_CSV
)


# ======================================================================
# STEP 52: AIS RESULT GRAPH
# ======================================================================

result_plot = ais_result_df.sort_values(

    by="Deviation_Percentage"

).reset_index(
    drop=True
)


plt.figure(
    figsize=(14, 9)
)


plt.barh(

    result_plot[
        "District"
    ],

    result_plot[
        "Deviation_Percentage"
    ]
)


plt.axvline(
    0,
    linewidth=1
)


plt.axvline(

    -20,

    linestyle="--",

    linewidth=1,

    label="Deficient Threshold (-20%)"
)


plt.axvline(

    20,

    linestyle="--",

    linewidth=1,

    label="Excess Threshold (+20%)"
)


plt.xlabel(
    "Rainfall Deviation (%)"
)


plt.ylabel(
    "District"
)


plt.title(
    "AIS District-wise Rainfall Deviation Analysis"
)


plt.grid(
    axis="x",
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    AIS_RESULT_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_RESULT_GRAPH
)


# ======================================================================
# STEP 53: FULL DATASET AIS PREDICTIONS
# ======================================================================

full_rf_prediction = rf_model.predict(
    X_full_scaled
)


full_rf_probability = rf_model.predict_proba(
    X_full_scaled
)


full_nn_probability = nn_model.predict(

    X_full_scaled,

    verbose=0
)


full_nn_prediction = np.argmax(

    full_nn_probability,

    axis=1
)


# ======================================================================
# STEP 54: DECODE PREDICTIONS
# ======================================================================

rf_prediction_names = (
    label_encoder.inverse_transform(
        full_rf_prediction
    )
)


nn_prediction_names = (
    label_encoder.inverse_transform(
        full_nn_prediction
    )
)


rf_confidence = np.max(

    full_rf_probability,

    axis=1

) * 100


nn_confidence = np.max(

    full_nn_probability,

    axis=1

) * 100


# ======================================================================
# STEP 55: CREATE AIS PREDICTION CSV
# ======================================================================

ais_prediction_df = pd.DataFrame({

    "District":

        df[
            district_column
        ].values,

    "Actual_Rainfall":

        df[
            "Calculated_Total_Actual"
        ].round(
            2
        ),

    "Normal_Rainfall":

        df[
            "Calculated_Total_Normal"
        ].round(
            2
        ),

    "Deviation_Percentage":

        df[
            "Rainfall_Deviation_Percentage"
        ].round(
            2
        ),

    "Actual_Category":

        df[
            "Rainfall_Category"
        ].values,

    "AIS_RF_Predicted_Category":

        rf_prediction_names,

    "AIS_RF_Confidence_Percentage":

        np.round(
            rf_confidence,
            2
        ),

    "AIS_NN_Predicted_Category":

        nn_prediction_names,

    "AIS_NN_Confidence_Percentage":

        np.round(
            nn_confidence,
            2
        )
})


ais_prediction_df.to_csv(

    AIS_PREDICTION_CSV,

    index=False
)


print(
    "\nSaved:",
    AIS_PREDICTION_CSV
)


print("\nAIS Predictions:")

print(
    ais_prediction_df
)


# ======================================================================
# STEP 56: AIS PREDICTION GRAPH
# ======================================================================

prediction_plot = (
    ais_prediction_df
    .sort_values(
        by="Deviation_Percentage"
    )
    .reset_index(
        drop=True
    )
)


x = np.arange(
    len(
        prediction_plot
    )
)


plt.figure(
    figsize=(16, 8)
)


plt.plot(

    x,

    prediction_plot[
        "AIS_RF_Confidence_Percentage"
    ],

    marker="o",

    label="AIS Random Forest Confidence"
)


plt.plot(

    x,

    prediction_plot[
        "AIS_NN_Confidence_Percentage"
    ],

    marker="s",

    label="AIS Neural Network Confidence"
)


plt.xticks(

    x,

    prediction_plot[
        "District"
    ],

    rotation=90
)


plt.xlabel(
    "District"
)


plt.ylabel(
    "Prediction Confidence (%)"
)


plt.title(
    "AIS District-wise Rainfall Prediction Confidence"
)


plt.ylim(
    0,
    105
)


plt.grid(
    alpha=0.3
)


plt.legend()


plt.tight_layout()


plt.savefig(

    AIS_PREDICTION_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_PREDICTION_GRAPH
)


# ======================================================================
# STEP 57: AIS ACTUAL VS NORMAL GRAPH
# ======================================================================

x = np.arange(
    len(
        ais_result_df
    )
)


width = 0.40


plt.figure(
    figsize=(16, 8)
)


plt.bar(

    x - width / 2,

    ais_result_df[
        "Actual_Rainfall"
    ],

    width,

    label="Actual Rainfall"
)


plt.bar(

    x + width / 2,

    ais_result_df[
        "Normal_Rainfall"
    ],

    width,

    label="Normal Rainfall"
)


plt.xticks(

    x,

    ais_result_df[
        "District"
    ],

    rotation=90
)


plt.xlabel(
    "District"
)


plt.ylabel(
    "Rainfall"
)


plt.title(
    "AIS Actual vs Normal Rainfall by District"
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    AIS_ACTUAL_NORMAL_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_ACTUAL_NORMAL_GRAPH
)


# ======================================================================
# STEP 58: AIS CATEGORY DISTRIBUTION GRAPH
# ======================================================================

category_counts = ais_result_df[

    "Rainfall_Category"

].value_counts()


plt.figure(
    figsize=(8, 6)
)


bars = plt.bar(

    category_counts.index,

    category_counts.values
)


plt.xlabel(
    "Rainfall Category"
)


plt.ylabel(
    "Number of Districts"
)


plt.title(
    "AIS Rainfall Category Distribution"
)


plt.grid(
    axis="y",
    alpha=0.3
)


for bar in bars:

    height = (
        bar.get_height()
    )


    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        height + 0.1,

        str(
            int(
                height
            )
        ),

        ha="center"
    )


plt.tight_layout()


plt.savefig(

    AIS_CATEGORY_GRAPH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    AIS_CATEGORY_GRAPH
)


# ======================================================================
# STEP 59: FEATURE IMPORTANCE AFTER AIS
# ======================================================================

feature_importance_path = os.path.join(

    PROJECT_DIR,

    "ais_feature_importance_graph.png"
)


importance_df = pd.DataFrame({

    "Feature":

        selected_features,

    "Importance":

        rf_model.feature_importances_
})


importance_df = importance_df.sort_values(

    by="Importance",

    ascending=True
)


plt.figure(
    figsize=(10, 7)
)


plt.barh(

    importance_df[
        "Feature"
    ],

    importance_df[
        "Importance"
    ]
)


plt.xlabel(
    "Feature Importance"
)


plt.ylabel(
    "AIS Selected Feature"
)


plt.title(
    "Random Forest Feature Importance After AIS Selection"
)


plt.grid(
    axis="x",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    feature_importance_path,

    dpi=300,

    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "Saved:",
    feature_importance_path
)


# ======================================================================
# STEP 60: TEST SAVED AIS PKL
# ======================================================================

print("\n" + "=" * 80)

print(
    "TESTING SAVED AIS PKL MODEL"
)

print("=" * 80)


with open(

    AIS_PKL_PATH,

    "rb"

) as file:

    loaded_data = pickle.load(
        file
    )


loaded_rf = loaded_data[
    "random_forest_model"
]


loaded_scaler = loaded_data[
    "scaler"
]


loaded_features = loaded_data[
    "ais_selected_features"
]


sample = df[
    loaded_features
].iloc[
    [0]
]


sample_scaled = loaded_scaler.transform(
    sample
)


sample_prediction = loaded_rf.predict(
    sample_scaled
)


sample_category = (
    label_encoder.inverse_transform(
        sample_prediction
    )[0]
)


print("\nSample District:")

print(
    df.iloc[0][
        district_column
    ]
)


print("\nAIS PKL Prediction:")

print(
    sample_category
)


# ======================================================================
# STEP 61: TEST SAVED AIS H5
# ======================================================================

print("\n" + "=" * 80)

print(
    "TESTING SAVED AIS H5 MODEL"
)

print("=" * 80)


loaded_nn = tf.keras.models.load_model(
    AIS_H5_PATH
)


sample_nn_probability = loaded_nn.predict(

    sample_scaled,

    verbose=0
)


sample_nn_prediction = np.argmax(

    sample_nn_probability,

    axis=1
)


sample_nn_category = (
    label_encoder.inverse_transform(
        sample_nn_prediction
    )[0]
)


print("\nAIS H5 Prediction:")

print(
    sample_nn_category
)


# ======================================================================
# STEP 62: FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 80)

print(
    "AIS PROJECT COMPLETED SUCCESSFULLY"
)

print("=" * 80)


print(
    "\nOriginal Features:",
    len(
        feature_columns
    )
)


print(
    "AIS Selected Features:",
    len(
        selected_features
    )
)


print(
    "AIS Best Fitness:",
    round(
        best_fitness,
        4
    )
)


print(
    f"\nAIS Random Forest Accuracy: "
    f"{rf_accuracy * 100:.2f}%"
)


print(
    f"AIS Neural Network Accuracy: "
    f"{nn_accuracy * 100:.2f}%"
)


print(
    "\nAIS Selected Features:"
)


for feature in selected_features:

    print(
        " -",
        feature
    )


# ======================================================================
# STEP 63: PRINT ALL GENERATED FILES
# ======================================================================

generated_files = [

    AIS_H5_PATH,

    AIS_PKL_PATH,

    AIS_YAML_PATH,

    AIS_JSON_PATH,

    AIS_ACCURACY_GRAPH,

    AIS_HEATMAP_GRAPH,

    AIS_COMPARISON_GRAPH,

    AIS_RESULT_CSV,

    AIS_RESULT_GRAPH,

    AIS_PREDICTION_CSV,

    AIS_PREDICTION_GRAPH,

    AIS_FEATURE_SELECTION_CSV,

    AIS_FEATURE_SELECTION_GRAPH,

    AIS_ACTUAL_NORMAL_GRAPH,

    AIS_CATEGORY_GRAPH,

    feature_importance_path
]


print("\nGenerated AIS Files:")


for number, path in enumerate(

    generated_files,

    start=1
):

    print(
        f"{number}. {path}"
    )


print("\n" + "=" * 80)

print(
    "ALL AIS OUTPUTS SAVED SUCCESSFULLY"
)

print("=" * 80)
```
